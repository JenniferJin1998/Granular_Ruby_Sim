## 1.3) Loading Direction Stress Distrituion

### 1.3.1) 0 Degree

```python
# -----------------------------
# Loading Direction Stress Distribution (0 Degree)
# -----------------------------

def load_fig_xy_export(mat_path: Path):
    d = loadmat(mat_path)
    return (
        np.asarray(d["sim_x"]).squeeze(),
        np.asarray(d["sim_y"]).squeeze(),
        np.asarray(d["exp_x"]).squeeze(),
        np.asarray(d["exp_y"]).squeeze(),
    )


def downsample_xy(x, y, step=30):
    """Keep every `step` points and always include the last point."""
    n = len(x)
    if step <= 1 or n <= step:
        return x, y
    idx = np.r_[np.arange(0, n, step), n - 1]
    idx = np.unique(idx)
    return x[idx], y[idx]


# Options: "filled_hist" or "scatter"
plot_mode = "scatter"

# Set to 1 to disable downsampling
downsample_step_sim = 8
downsample_step_exp = 12

# -----------------------------
# Global font settings
# -----------------------------
font_size = 24
plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["font.size"] = font_size

# -----------------------------
# Figure and axes settings
# -----------------------------
figsize = (8, 6)
axis_label_size = 28
tick_label_size = 24
spine_width = 1
tick_width = 1
tick_length = 5

# -----------------------------
# Style controls
# -----------------------------
sim_fill = {
    "color": "#00274C",
    "fill_alpha": 0.35,
    "line_alpha": 1.0,
    "line_width": 2.2,
    "line_style": "-",
}

exp_fill = {
    "color": "#FFCB05",
    "fill_alpha": 0.35,
    "line_alpha": 1.0,
    "line_width": 2.2,
    "line_style": "--",
}

sim_scatter = {
    "marker": "o",
    "color": "#00274C",
    "size": 90,
    "edge_color": "white",
    "edge_width": 0.9,
}

exp_scatter = {
    "marker": "s",
    "color": "#FFCB05",
    "size": 90,
    "edge_color": "black",
    "edge_width": 0.9,
}

show_connecting_lines = True

xy_mat = Path("/Users/yfjin/Desktop/PhD/Presentation/TMS2026/Granular_Figures/PlotData/0Degree/Config0_Sam1_zzStress_normalized_curve_xy.mat")
sim_x, sim_y, exp_x, exp_y = load_fig_xy_export(xy_mat)

sim_x_ds, sim_y_ds = downsample_xy(sim_x, sim_y, step=downsample_step_sim)
exp_x_ds, exp_y_ds = downsample_xy(exp_x, exp_y, step=downsample_step_exp)

fig, ax = plt.subplots(figsize=figsize)

if plot_mode == "filled_hist":
    ax.fill_between(sim_x_ds, sim_y_ds, step="mid", color=sim_fill["color"], alpha=sim_fill["fill_alpha"])
    ax.fill_between(exp_x_ds, exp_y_ds, step="mid", color=exp_fill["color"], alpha=exp_fill["fill_alpha"])

    ax.plot(sim_x_ds, sim_y_ds, color=sim_fill["color"], alpha=sim_fill["line_alpha"], linewidth=sim_fill["line_width"], linestyle=sim_fill["line_style"])
    ax.plot(exp_x_ds, exp_y_ds, color=exp_fill["color"], alpha=exp_fill["line_alpha"], linewidth=exp_fill["line_width"], linestyle=exp_fill["line_style"])

elif plot_mode == "scatter":
    if show_connecting_lines:
        ax.plot(sim_x_ds, sim_y_ds, color=sim_fill["color"], linewidth=1.6, linestyle="-", alpha=0.9)
        ax.plot(exp_x_ds, exp_y_ds, color=exp_fill["color"], linewidth=1.6, linestyle="--", alpha=0.95)

    ax.scatter(
        sim_x_ds,
        sim_y_ds,
        marker=sim_scatter["marker"],
        s=sim_scatter["size"],
        facecolor=sim_scatter["color"],
        edgecolor=sim_scatter["edge_color"],
        linewidth=sim_scatter["edge_width"],
        zorder=3,
    )
    ax.scatter(
        exp_x_ds,
        exp_y_ds,
        marker=exp_scatter["marker"],
        s=exp_scatter["size"],
        facecolor=exp_scatter["color"],
        edgecolor=exp_scatter["edge_color"],
        linewidth=exp_scatter["edge_width"],
        zorder=3,
    )

else:
    raise ValueError("plot_mode must be 'filled_hist' or 'scatter'")

ax.set_title("0 Degree", fontsize=axis_label_size)
ax.set_xlabel("Loading Direction Stress (Normalized)", fontsize=axis_label_size)
ax.set_ylabel("PDF", fontsize=axis_label_size)
ax.tick_params(axis="both", which="major", labelsize=tick_label_size, width=tick_width, length=tick_length)

for spine in ax.spines.values():
    spine.set_linewidth(spine_width)

fig.tight_layout()
plt.show()
```

### 2.3.3) Displacement vs Force across Geometries
Convert φ → displacement using container geometry.  
φ = V_particles / (A_container · H)  →  H = V_particles / (A · φ),  disp = H₀ − H

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat

# ---- physical parameters ----
d_particle   = 1.5e-4   # m  (particle diameter)
N_particles  = 464
D_container  = 1.5e-3   # m  (container diameter)

V_particles = N_particles * (np.pi / 6) * d_particle**3   # total particle volume [m³]
A_container = np.pi * (D_container / 2)**2                 # cross-sectional area [m²]

def phi_to_height(phi: np.ndarray) -> np.ndarray:
    """Container height at each timestep from packing density."""
    return V_particles / (A_container * phi)

def phi_to_displacement_mm(phi: np.ndarray) -> np.ndarray:
    """Displacement in mm, zero-aligned to the first timestep."""
    H = phi_to_height(phi)
    return (H[0] - H) * 1e3   # m → mm

# ---- load φ arrays ----
plot_data_dir = Path("/Users/yfjin/Desktop/PhD/Presentation/TMS2026/Granular_Figures/PlotData")
angles = [0, 15, 30, 45]

def load_phi(mat_path: Path) -> np.ndarray:
    data = loadmat(mat_path)
    keys = [k for k in data.keys() if not k.startswith("__")]
    return np.asarray(data[keys[0]]).squeeze()

phi_arrays = [load_phi(plot_data_dir / f"phi_list_{a}.mat") for a in angles]

# ---- force axis: 0.5 N → 29.5 N ----
force_axis = np.arange(0.5, 30.0, 1.0)

# ---- style ----
plt.rcParams["font.family"] = "Times New Roman"
figsize         = (6, 5)
axis_label_size = 20
tick_label_size = 16
spine_width     = 1
tick_width      = 1
tick_length     = 5

geo_styles = {
    0:  {"color": "#00274C", "marker": "o", "label": "0°"},
    15: {"color": "#C0392B", "marker": "s", "label": "15°"},
    30: {"color": "#27AE60", "marker": "^", "label": "30°"},
    45: {"color": "#8E44AD", "marker": "D", "label": "45°"},
}

# ---- plot ----
fig, ax = plt.subplots(figsize=figsize)

for ang, phi in zip(angles, phi_arrays):
    sty  = geo_styles[ang]
    disp = phi_to_displacement_mm(phi)
    ax.plot(
        disp,
        force_axis,
        color=sty["color"],
        linestyle="-",
        linewidth=2,
        marker=sty["marker"],
        markersize=6,
        markerfacecolor=sty["color"],
        markeredgecolor="white",
        markeredgewidth=0.6,
        label=sty["label"],
        markevery=4,
    )

ax.set_xlabel("Displacement (mm)", fontsize=axis_label_size)
ax.set_ylabel("Force (N)", fontsize=axis_label_size)
ax.tick_params(
    axis="both", which="major",
    labelsize=tick_label_size, width=tick_width, length=tick_length,
)
for spine in ax.spines.values():
    spine.set_linewidth(spine_width)

ax.legend(fontsize=tick_label_size, frameon=False)
fig.tight_layout()
plt.show()

# ---- print displacement range per geometry ----
for ang, phi in zip(angles, phi_arrays):
    disp = phi_to_displacement_mm(phi)
    print(f"{ang}°:  0.000 → {disp[-1]:.4f} mm  (total {disp[-1]:.4f} mm)")
```

```python
# phi_arrays, force_axis, angles, geo_styles, and style params are set in §2.3/2.4 cells above.
# strain = (H0 - H) / H0 = 1 - phi[0] / phi_i  (dimensionless, expressed as %)

plt.rcParams["font.family"] = "Times New Roman"

fig, ax = plt.subplots(figsize=figsize)

for ang, phi in zip(angles, phi_arrays):
    sty    = geo_styles[ang]
    strain = (1.0 - phi[0] / phi) * 100.0   # percent
    ax.plot(
        strain,
        force_axis,
        color=sty["color"],
        linestyle="-",
        linewidth=2,
        marker=sty["marker"],
        markersize=6,
        markerfacecolor=sty["color"],
        markeredgecolor="white",
        markeredgewidth=0.6,
        label=sty["label"],
        markevery=4,
    )

ax.set_xlabel("Strain (%)", fontsize=axis_label_size)
ax.set_ylabel("Force (N)", fontsize=axis_label_size)
ax.tick_params(
    axis="both", which="major",
    labelsize=tick_label_size, width=tick_width, length=tick_length,
)
for spine in ax.spines.values():
    spine.set_linewidth(spine_width)

ax.legend(fontsize=tick_label_size, frameon=False)
fig.tight_layout()
plt.show()

for ang, phi in zip(angles, phi_arrays):
    strain = (1.0 - phi[0] / phi) * 100.0
    print(f"{ang}°:  0.000 → {strain[-1]:.4f} %")
```

```python
from pathlib import Path
from zipfile import ZipFile
import xml.etree.ElementTree as ET
import re

import numpy as np
import matplotlib.pyplot as plt

# ---- workbook location ----
xlsx_path = Path("/Users/yfjin/Desktop/PhD/CHESS02142024/GraphNetwork/Experiment/ForceDisplacement.xlsx")

# ---- XML helpers for .xlsx parsing ----
sheet_ns = {"a": "http://schemas.openxmlformats.org/spreadsheetml/2006/main"}
chart_ns = {"c": "http://schemas.openxmlformats.org/drawingml/2006/chart"}
cell_ref_re = re.compile(r"^\$?([A-Z]+)\$?(\d+)$")
range_re = re.compile(r"^(?:'([^']+)'|([^!]+)|)(?:!)?\$?([A-Z]+)\$?(\d+):\$?([A-Z]+)\$?(\d+)$")


def col_to_index(col_letters: str) -> int:
    value = 0
    for char in col_letters:
        value = value * 26 + (ord(char) - ord("A") + 1)
    return value


def index_to_col(col_idx: int) -> str:
    letters = []
    while col_idx > 0:
        col_idx, rem = divmod(col_idx - 1, 26)
        letters.append(chr(ord("A") + rem))
    return "".join(reversed(letters))


def parse_cell_ref(cell_ref: str):
    match = cell_ref_re.match(cell_ref)
    if not match:
        raise ValueError(f"Invalid cell reference: {cell_ref}")
    col_letters, row_str = match.groups()
    return col_to_index(col_letters), int(row_str)


def parse_range(formula: str):
    match = range_re.match(formula)
    if not match:
        raise ValueError(f"Unsupported range formula: {formula}")
    sheet_a, sheet_b, start_col, start_row, end_col, end_row = match.groups()
    return {
        "sheet": sheet_a or sheet_b,
        "start_col": col_to_index(start_col),
        "start_row": int(start_row),
        "end_col": col_to_index(end_col),
        "end_row": int(end_row),
    }


def load_shared_strings(zip_file: ZipFile):
    if "xl/sharedStrings.xml" not in zip_file.namelist():
        return []

    root = ET.fromstring(zip_file.read("xl/sharedStrings.xml"))
    strings = []
    for si in root.findall("a:si", sheet_ns):
        parts = [node.text or "" for node in si.iterfind(".//a:t", sheet_ns)]
        strings.append("".join(parts))
    return strings


def load_sheet_cells(zip_file: ZipFile, sheet_path: str, shared_strings):
    root = ET.fromstring(zip_file.read(sheet_path))
    cells = {}

    for row in root.find("a:sheetData", sheet_ns).findall("a:row", sheet_ns):
        for cell in row.findall("a:c", sheet_ns):
            ref = cell.attrib["r"]
            value_node = cell.find("a:v", sheet_ns)
            if value_node is None:
                continue

            raw_value = value_node.text
            if cell.attrib.get("t") == "s":
                value = shared_strings[int(raw_value)]
            else:
                value = float(raw_value)
            cells[ref] = value

    return cells


def extract_chart_series(zip_file: ZipFile):
    chart_root = ET.fromstring(zip_file.read("xl/charts/chart1.xml"))
    specs = []
    for ser in chart_root.findall(".//c:scatterChart/c:ser", chart_ns):
        x_formula = ser.findtext("c:xVal/c:numRef/c:f", namespaces=chart_ns)
        y_formula = ser.findtext("c:yVal/c:numRef/c:f", namespaces=chart_ns)
        if x_formula and y_formula:
            specs.append({"x_formula": x_formula, "y_formula": y_formula})
    return specs


def get_range_values(cells, range_info):
    if range_info["start_col"] != range_info["end_col"]:
        raise ValueError("Only single-column ranges are supported.")

    col_letters = index_to_col(range_info["start_col"])
    values = []
    for row_idx in range(range_info["start_row"], range_info["end_row"] + 1):
        cell_ref = f"{col_letters}{row_idx}"
        if cell_ref in cells:
            values.append(float(cells[cell_ref]))
    return np.asarray(values, dtype=float)


def find_block_label(cells, data_col_idx, label_row=1):
    label = None
    label_col = None
    for cell_ref, value in cells.items():
        col_idx, row_idx = parse_cell_ref(cell_ref)
        if row_idx != label_row or not isinstance(value, str):
            continue
        if col_idx <= data_col_idx and (label_col is None or col_idx > label_col):
            label = value
            label_col = col_idx
    return label


config_to_angle = {
    "Config0_Sam1": 0,
    "Config2_Sam2": 30,
    "Config4_Sam1": 60,
}

exp_styles = {
    0:  {"color": "#00274C", "marker": "o", "label": "0°"},
    30: {"color": "#27AE60", "marker": "^", "label": "30°"},
    60: {"color": "#C0392B", "marker": "s", "label": "60°"},
}

# ---- load experimental series from the workbook/chart ----
with ZipFile(xlsx_path) as zf:
    shared_strings = load_shared_strings(zf)
    sheet_cells = load_sheet_cells(zf, "xl/worksheets/sheet1.xml", shared_strings)
    chart_specs = extract_chart_series(zf)

exp_series = []
for spec in chart_specs:
    x_range = parse_range(spec["x_formula"])
    y_range = parse_range(spec["y_formula"])
    config_label = find_block_label(sheet_cells, min(x_range["start_col"], y_range["start_col"]))
    angle = config_to_angle[config_label]
    displacement = get_range_values(sheet_cells, x_range)
    force = get_range_values(sheet_cells, y_range)
    n = min(len(displacement), len(force))
    displacement = displacement[:n] - displacement[0]
    force = force[:n] - force[0]
    exp_series.append({
        "config_label": config_label,
        "angle": angle,
        "displacement": displacement,
        "force": force,
    })

exp_series = sorted(exp_series, key=lambda item: item["angle"])

# ---- plot ----
axis_label_size = globals().get("axis_label_size", 20)
tick_label_size = globals().get("tick_label_size", 16)
spine_width = globals().get("spine_width", 1)
tick_width = globals().get("tick_width", 1)
tick_length = globals().get("tick_length", 5)

plt.rcParams["font.family"] = "Times New Roman"
exp_figsize = (6, 5)

fig, ax = plt.subplots(figsize=exp_figsize)

for series in exp_series:
    sty = exp_styles[series["angle"]]
    ax.plot(
        series["displacement"],
        series["force"],
        color=sty["color"],
        linestyle="-",
        linewidth=2,
        marker=sty["marker"],
        markersize=5,
        markerfacecolor=sty["color"],
        markeredgecolor="white",
        markeredgewidth=0.6,
        label=sty["label"],
        markevery=max(1, len(series["displacement"]) // 12),
    )

ax.set_xlabel("Displacement", fontsize=axis_label_size)
ax.set_ylabel("Force (N)", fontsize=axis_label_size)
ax.tick_params(
    axis="both", which="major",
    labelsize=tick_label_size, width=tick_width, length=tick_length,
)
for spine in ax.spines.values():
    spine.set_linewidth(spine_width)

ax.legend(fontsize=tick_label_size, frameon=False)
fig.tight_layout()
plt.show()

for series in exp_series:
    print(
        f"{series['config_label']} ({series['angle']}°):  "
        f"disp {series['displacement'][0]:.4f} → {series['displacement'][-1]:.4f},  "
        f"force {series['force'][0]:.4f} → {series['force'][-1]:.4f}"
    )
```

## 2.4) Bin Plot of Properties

### 2.4.0) Functions for Bin Plots

```python
from typing import Dict, Optional, Tuple
from pathlib import Path
import math
import pickle
import re

import matplotlib.pyplot as plt
import numpy as np


def _angle_from_label(label: str) -> float:
    """Parse geometry angle from labels like '0deg', '15deg', 'Degree30'."""
    s = str(label).lower()
    if s.endswith("deg"):
        return float(s[:-3])
    m = re.search(r"(\d+(?:\.\d+)?)", s)
    if m:
        return float(m.group(1))
    raise ValueError(f"Cannot parse angle from geometry label: {label}")


def _sorted_geometry_labels(labels):
    return sorted(labels, key=_angle_from_label)


def _pick_graph_list(entry, graph_view: str = "core"):
    """Accept multiple graph_dict layouts and return list of graphs."""
    if isinstance(entry, dict):
        if graph_view in entry and isinstance(entry[graph_view], (list, tuple)):
            return entry[graph_view]
        for value in entry.values():
            if isinstance(value, (list, tuple)):
                return value
        raise ValueError("Geometry entry dict does not contain a graph list")
    if isinstance(entry, (list, tuple)):
        return entry
    raise TypeError("Unsupported graph_dict geometry entry type")


def _node_value_from_property(node_data: dict, property_name: str) -> float:
    """Read any numeric node property, with stress aliases and case-insensitive fallback."""
    key = property_name.lower()
    alias_map = {
        "hydro": ["stress_hydro", "hydro", "hydrostatic"],
        "hydrostatic": ["stress_hydro", "hydro", "hydrostatic"],
        "vm": ["stress_vm", "vm", "von_mises", "vonmises"],
        "von_mises": ["stress_vm", "vm", "von_mises", "vonmises"],
        "vonmises": ["stress_vm", "vm", "von_mises", "vonmises"],
        "s11": ["stress_11", "sigma11", "s11"],
        "s22": ["stress_22", "sigma22", "s22"],
        "s33": ["stress_33", "sigma33", "s33"],
        "s12": ["stress_12", "sigma12", "s12"],
        "s13": ["stress_13", "sigma13", "s13"],
        "s23": ["stress_23", "sigma23", "s23"],
    }
    keys = list(alias_map.get(key, [property_name]))

    if property_name not in keys:
        keys.append(property_name)
    if key not in keys:
        keys.append(key)

    lower_to_actual = {str(k).lower(): k for k in node_data.keys()}
    for k in keys:
        actual_key = k if k in node_data else lower_to_actual.get(str(k).lower(), None)
        if actual_key is None:
            continue
        val = node_data.get(actual_key)
        if np.isscalar(val):
            try:
                return float(val)
            except (TypeError, ValueError):
                pass
    return np.nan


def load_sim_particle_data_from_graph_dict(
    graph_dict_path: Path,
    property_name: str = "hydro",
    graph_view: str = "core",
    geometry_labels: Optional[list] = None,
) -> Dict[str, Tuple[np.ndarray, np.ndarray]]:
    """Load per-geometry particle positions and node property from graph_dict pickle."""
    with open(graph_dict_path, "rb") as f:
        graph_dict = pickle.load(f)

    if not isinstance(graph_dict, dict):
        raise TypeError("Loaded graph_dict is not a dict")

    labels = geometry_labels if geometry_labels is not None else list(graph_dict.keys())
    labels = _sorted_geometry_labels(labels)

    out: Dict[str, Tuple[np.ndarray, np.ndarray]] = {}

    for geom in labels:
        if geom not in graph_dict:
            continue
        graphs = _pick_graph_list(graph_dict[geom], graph_view=graph_view)

        positions = []
        values = []

        for graph in graphs:
            for _, node_data in graph.nodes(data=True):
                if node_data.get("is_wall", False):
                    continue
                pos = node_data.get("position", None)
                if pos is None:
                    continue
                pos = np.asarray(pos, dtype=float).reshape(-1)
                if pos.size < 3:
                    continue
                val = _node_value_from_property(node_data, property_name=property_name)
                if np.isfinite(val):
                    positions.append(pos[:3])
                    values.append(val)

        if len(positions) == 0:
            out[geom] = (np.empty((0, 3), dtype=float), np.empty((0,), dtype=float))
        else:
            out[geom] = (np.asarray(positions, dtype=float), np.asarray(values, dtype=float))

    return out


def _z_eff_from_rz(
    r: np.ndarray,
    z: np.ndarray,
    slope_k: float,
    use_dual_slope: bool = False,
    split_r: Optional[float] = None,
) -> np.ndarray:
    """Transform z to z_eff for bin indexing."""
    if not use_dual_slope:
        return z - slope_k * r

    if split_r is None:
        split_r = float(np.nanmedian(r))

    left = z - slope_k * r
    right = z + slope_k * r - 2.0 * slope_k * split_r
    return np.where(r <= split_r, left, right)


def _build_inclined_mesh(
    r_edges: np.ndarray,
    z_eff_edges: np.ndarray,
    slope_k: float,
    use_dual_slope: bool = False,
    split_r: Optional[float] = None,
):
    """Build corner mesh in physical (r, z) from transformed coordinates."""
    r_grid = np.repeat(r_edges[None, :], len(z_eff_edges), axis=0)
    z_eff_grid = np.repeat(z_eff_edges[:, None], len(r_edges), axis=1)

    if not use_dual_slope:
        z_grid = z_eff_grid + slope_k * r_grid
        return r_grid, z_grid

    if split_r is None:
        split_r = float(np.nanmedian(r_edges))

    left = z_eff_grid + slope_k * r_grid
    right = z_eff_grid - slope_k * r_grid + 2.0 * slope_k * split_r
    z_grid = np.where(r_grid <= split_r, left, right)
    return r_grid, z_grid


def bin_particles_rz(
    positions_xyz: np.ndarray,
    values: np.ndarray,
    angle_deg: float,
    n_r_bins: int = 13,
    n_z_bins: int = 12,
    center_xyz: Optional[np.ndarray] = None,
    align_with_boundary_shape: bool = True,
    slope_sign: int = 1,
    use_dual_slope: bool = False,
    split_r: Optional[float] = None,
) -> Dict[str, np.ndarray]:
    """
    Bin particles in (r, z) with optional inclined cells.

    Single-slope mode: z_eff = z - k*r.
    Dual-slope mode:   z_eff = z - k*r               for r <= r_split
                       z_eff = z + k*r - 2*k*r_split for r >  r_split
    where k = slope_sign * tan(angle_deg).
    """
    if positions_xyz.size == 0 or values.size == 0:
        return {
            "avg": np.full((n_z_bins, n_r_bins), np.nan),
            "counts": np.zeros((n_z_bins, n_r_bins), dtype=int),
            "r_edges": np.linspace(0.0, 1.0, n_r_bins + 1),
            "z_eff_edges": np.linspace(0.0, 1.0, n_z_bins + 1),
            "r_mesh": None,
            "z_mesh": None,
            "center_xyz": np.array([np.nan, np.nan, np.nan]),
            "incline_k": 0.0,
            "split_r": np.nan,
        }

    if center_xyz is None:
        center_xyz = positions_xyz.mean(axis=0)
    center_xyz = np.asarray(center_xyz, dtype=float).reshape(1, 3)

    centered = positions_xyz - center_xyz
    r = np.sqrt(centered[:, 0] ** 2 + centered[:, 1] ** 2)
    z = centered[:, 2]
    z = z - np.nanmin(z)

    k = 0.0
    if align_with_boundary_shape and abs(angle_deg) > 1e-12:
        k = slope_sign * math.tan(math.radians(angle_deg))

    split_r_used = float(np.nanmedian(r)) if split_r is None else float(split_r)
    z_eff = _z_eff_from_rz(r, z, slope_k=k, use_dual_slope=use_dual_slope, split_r=split_r_used)

    r_min, r_max = float(np.nanmin(r)), float(np.nanmax(r))
    z_eff_min, z_eff_max = float(np.nanmin(z_eff)), float(np.nanmax(z_eff))
    if r_min == r_max:
        r_max = r_min + 1e-12
    if z_eff_min == z_eff_max:
        z_eff_max = z_eff_min + 1e-12

    r_edges = np.linspace(r_min, r_max, n_r_bins + 1)
    z_eff_edges = np.linspace(z_eff_min, z_eff_max, n_z_bins + 1)

    r_idx = np.digitize(r, r_edges) - 1
    z_idx = np.digitize(z_eff, z_eff_edges) - 1
    r_idx = np.clip(r_idx, 0, n_r_bins - 1)
    z_idx = np.clip(z_idx, 0, n_z_bins - 1)

    sums = np.zeros((n_z_bins, n_r_bins), dtype=float)
    counts = np.zeros((n_z_bins, n_r_bins), dtype=int)
    np.add.at(sums, (z_idx, r_idx), values)
    np.add.at(counts, (z_idx, r_idx), 1)

    avg = np.full_like(sums, np.nan, dtype=float)
    valid = counts > 0
    avg[valid] = sums[valid] / counts[valid]

    r_mesh, z_mesh = _build_inclined_mesh(
        r_edges,
        z_eff_edges,
        slope_k=k,
        use_dual_slope=use_dual_slope,
        split_r=split_r_used,
    )

    return {
        "avg": avg,
        "counts": counts,
        "r_edges": r_edges,
        "z_eff_edges": z_eff_edges,
        "r_mesh": r_mesh,
        "z_mesh": z_mesh,
        "center_xyz": center_xyz.ravel(),
        "incline_k": k,
        "split_r": split_r_used,
    }


def _mean_std_bounds(arr: np.ndarray, std_factor: float) -> Tuple[float, float]:
    finite = arr[np.isfinite(arr)]
    if finite.size == 0:
        return -1.0, 1.0

    mu = float(np.nanmean(finite))
    sd = float(np.nanstd(finite))
    if sd <= 0:
        sd = max(abs(mu) * 1e-6, 1e-6)
    return mu - std_factor * sd, mu + std_factor * sd


def _compute_color_scales(
    maps_by_geom: Dict[str, np.ndarray],
    color_range_mode: str = "mean_std_per_geometry",
    std_factor: float = 2.0,
    global_minmax: Optional[Tuple[float, float]] = None,
) -> Dict[str, Tuple[float, float]]:
    """
    Two explicit color-range modes only:
    1) mean_std_per_geometry: each geometry uses mean +/- n*std (default n=2)
    2) custom_global_minmax: all geometries share user-provided (vmin, vmax)
    """
    if color_range_mode == "mean_std_per_geometry":
        return {geom: _mean_std_bounds(arr, std_factor=std_factor) for geom, arr in maps_by_geom.items()}

    if color_range_mode == "custom_global_minmax":
        if global_minmax is None:
            raise ValueError(
                "global_minmax must be provided for color_range_mode='custom_global_minmax'"
            )
        vmin, vmax = float(global_minmax[0]), float(global_minmax[1])
        if vmin >= vmax:
            raise ValueError("global_minmax must satisfy vmin < vmax")
        return {geom: (vmin, vmax) for geom in maps_by_geom}

    raise ValueError(
        "color_range_mode must be 'mean_std_per_geometry' or 'custom_global_minmax'"
    )


def create_simulation_rz_binplots_from_graph_dict(
    graph_dict_path: Path,
    property_name: str = "hydro",
    property_sign: int = 1,
    graph_view: str = "core",
    geometry_labels: Optional[list] = None,
    n_r_bins: int = 13,
    n_z_bins: int = 12,
    center_by_geometry: Optional[Dict[str, np.ndarray]] = None,
    color_range_mode: str = "mean_std_per_geometry",
    std_factor: float = 2.0,
    global_minmax: Optional[Tuple[float, float]] = None,
    align_with_boundary_shape: bool = True,
    slope_sign: int = 1,
    use_dual_slope: bool = False,
    split_r_by_geometry: Optional[Dict[str, float]] = None,
    invert_z_plot: bool = False,
    cmap: str = "RdBu_r",
) -> Tuple[plt.Figure, Dict[str, Dict[str, np.ndarray]], Dict[str, Tuple[float, float]]]:
    """Create geometry bin plots in (r, z) from graph_dict node data."""
    center_by_geometry = center_by_geometry or {}
    split_r_by_geometry = split_r_by_geometry or {}

    if property_sign not in (-1, 1):
        raise ValueError("property_sign must be either +1 or -1")

    geom_data = load_sim_particle_data_from_graph_dict(
        graph_dict_path=graph_dict_path,
        property_name=property_name,
        graph_view=graph_view,
        geometry_labels=geometry_labels,
    )

    out: Dict[str, Dict[str, np.ndarray]] = {}
    for geom_label in _sorted_geometry_labels(list(geom_data.keys())):
        pos, values = geom_data[geom_label]
        values = property_sign * values
        angle = _angle_from_label(geom_label)
        out[geom_label] = bin_particles_rz(
            positions_xyz=pos,
            values=values,
            angle_deg=angle,
            n_r_bins=n_r_bins,
            n_z_bins=n_z_bins,
            center_xyz=center_by_geometry.get(geom_label),
            align_with_boundary_shape=align_with_boundary_shape,
            slope_sign=slope_sign,
            use_dual_slope=use_dual_slope,
            split_r=split_r_by_geometry.get(geom_label),
        )

    maps = {geom: data["avg"] for geom, data in out.items()}
    color_scales = _compute_color_scales(
        maps_by_geom=maps,
        color_range_mode=color_range_mode,
        std_factor=std_factor,
        global_minmax=global_minmax,
    )

    n = len(out)
    ncols = 2
    nrows = max(1, int(np.ceil(n / ncols)))
    fig, axes = plt.subplots(nrows, ncols, figsize=(14, 5 * nrows), constrained_layout=True)
    axes = np.atleast_1d(axes).ravel()

    for i, (geom_label, data) in enumerate(out.items()):
        ax = axes[i]
        mesh = ax.pcolormesh(
            data["r_mesh"],
            data["z_mesh"],
            data["avg"],
            shading="auto",
            cmap=cmap,
            vmin=color_scales[geom_label][0],
            vmax=color_scales[geom_label][1],
        )
        mode_tag = "dual" if use_dual_slope else "single"
        ax.set_title(
            f"{geom_label} | {property_name} x({property_sign:+d}) | k={data['incline_k']:.3f} | {mode_tag} | r_split={data['split_r']:.3e}",
            fontsize=10,
        )
        ax.set_xlabel("r")
        ax.set_ylabel("z (bottom-origin)")
        if invert_z_plot:
            ax.invert_yaxis()
            ax.set_ylabel("z (inverted for display)")
        cbar = plt.colorbar(mesh, ax=ax, pad=0.01)
        cbar.set_label(property_name)

    for j in range(len(out), len(axes)):
        axes[j].axis("off")

    return fig, out, color_scales
```

### 2.4.99) Experimental Property Bin Plots (Load-Step Based)

```python
# ------------------ 2.4.9 Experimental Property Bin Plots (Load-Step Based) ------------------


def _loadstep_sort_key(label):
    s = str(label)
    m = re.search(r"(\d+)", s)
    if m:
        return (0, int(m.group(1)), s)
    return (1, s.lower())


def _is_graph_like(obj) -> bool:
    return hasattr(obj, "nodes") and hasattr(obj, "edges")


def _collect_graphs_from_entry(entry, graph_view: str = "core"):
    """Recursively collect graph objects from mixed dict/list experiment layouts."""
    if _is_graph_like(entry):
        return [entry]

    if isinstance(entry, (list, tuple)):
        graphs = []
        for item in entry:
            graphs.extend(_collect_graphs_from_entry(item, graph_view=graph_view))
        return graphs

    if isinstance(entry, dict):
        if graph_view in entry:
            preferred = _collect_graphs_from_entry(entry[graph_view], graph_view=graph_view)
            if preferred:
                return preferred

        for key in ("graph", "graphs", "core", "full", "core_graph", "full_graph"):
            if key in entry:
                found = _collect_graphs_from_entry(entry[key], graph_view=graph_view)
                if found:
                    return found

        graphs = []
        for value in entry.values():
            graphs.extend(_collect_graphs_from_entry(value, graph_view=graph_view))
        return graphs

    return []


def _available_numeric_node_properties(graphs, max_nodes: int = 300):
    props = set()
    seen = 0
    for graph in graphs:
        try:
            node_iter = graph.nodes(data=True)
        except Exception:
            continue
        for _, node_data in node_iter:
            if node_data.get("is_wall", False):
                continue
            seen += 1
            for key, value in node_data.items():
                if key in ("position", "is_wall"):
                    continue
                if np.isscalar(value):
                    try:
                        _ = float(value)
                        props.add(str(key))
                    except (TypeError, ValueError):
                        pass
            if seen >= max_nodes:
                break
        if seen >= max_nodes:
            break
    return sorted(props)


def load_experiment_particle_data_from_graph_dict(
    graph_dict_path: Path,
    property_name: str = "avg_neighbor_degree",
    graph_view: str = "core",
    load_step_labels: Optional[list] = None,
) -> Dict[str, Tuple[np.ndarray, np.ndarray]]:
    """Load per-load-step particle positions and node property from experiment graph_dict pickle."""
    with open(graph_dict_path, "rb") as f:
        graph_dict = pickle.load(f)

    if not isinstance(graph_dict, dict):
        raise TypeError("Loaded graph_dict is not a dict")

    labels = load_step_labels if load_step_labels is not None else list(graph_dict.keys())
    labels = sorted(labels, key=_loadstep_sort_key)

    out: Dict[str, Tuple[np.ndarray, np.ndarray]] = {}
    all_graphs = []
    total_finite_values = 0

    for step in labels:
        if step not in graph_dict:
            continue

        graphs = _collect_graphs_from_entry(graph_dict[step], graph_view=graph_view)
        all_graphs.extend(graphs)

        if len(graphs) == 0:
            out[step] = (np.empty((0, 3), dtype=float), np.empty((0,), dtype=float))
            continue

        positions = []
        values = []

        for graph in graphs:
            try:
                node_iter = graph.nodes(data=True)
            except Exception:
                continue

            for _, node_data in node_iter:
                if node_data.get("is_wall", False):
                    continue
                pos = node_data.get("position", None)
                if pos is None:
                    continue
                pos = np.asarray(pos, dtype=float).reshape(-1)
                if pos.size < 3:
                    continue

                val = _node_value_from_property(node_data, property_name=property_name)
                if np.isfinite(val):
                    positions.append(pos[:3])
                    values.append(val)

        if len(positions) == 0:
            out[step] = (np.empty((0, 3), dtype=float), np.empty((0,), dtype=float))
        else:
            arr_pos = np.asarray(positions, dtype=float)
            arr_val = np.asarray(values, dtype=float)
            out[step] = (arr_pos, arr_val)
            total_finite_values += int(arr_val.size)

    if total_finite_values == 0:
        available = _available_numeric_node_properties(all_graphs)
        preview = ", ".join(available[:12]) if available else "<none detected>"
        raise ValueError(
            f"No finite node values found for property '{property_name}'. "
            f"Available numeric node properties include: {preview}"
        )

    return out


def create_experiment_rz_binplots_from_graph_dict(
    graph_dict_path: Path,
    property_name: str = "avg_neighbor_degree",
    property_sign: int = 1,
    graph_view: str = "core",
    load_step_labels: Optional[list] = None,
    angle_deg: float = 0.0,
    n_r_bins: int = 13,
    n_z_bins: int = 10,
    center_by_load_step: Optional[Dict[str, np.ndarray]] = None,
    color_range_mode: str = "mean_std_per_geometry",
    std_factor: float = 2.0,
    global_minmax: Optional[Tuple[float, float]] = None,
    align_with_boundary_shape: bool = True,
    slope_sign: int = 1,
    use_dual_slope: bool = False,
    split_r_by_load_step: Optional[Dict[str, float]] = None,
    invert_z_plot: bool = False,
    cmap: str = "RdBu_r",
) -> Tuple[plt.Figure, Dict[str, Dict[str, np.ndarray]], Dict[str, Tuple[float, float]]]:
    """
    Experimental version of the bin plot function.
    Same workflow as simulation, but keys are load steps instead of geometry labels.
    """
    center_by_load_step = center_by_load_step or {}
    split_r_by_load_step = split_r_by_load_step or {}

    if property_sign not in (-1, 1):
        raise ValueError("property_sign must be either +1 or -1")

    step_data = load_experiment_particle_data_from_graph_dict(
        graph_dict_path=graph_dict_path,
        property_name=property_name,
        graph_view=graph_view,
        load_step_labels=load_step_labels,
    )

    out: Dict[str, Dict[str, np.ndarray]] = {}
    for step in sorted(step_data.keys(), key=_loadstep_sort_key):
        pos, values = step_data[step]
        if pos.size > 0 and values.size > 0:
            finite_mask = np.all(np.isfinite(pos), axis=1) & np.isfinite(values)
            pos = pos[finite_mask]
            values = values[finite_mask]

        values = property_sign * values
        out[step] = bin_particles_rz(
            positions_xyz=pos,
            values=values,
            angle_deg=angle_deg,
            n_r_bins=n_r_bins,
            n_z_bins=n_z_bins,
            center_xyz=center_by_load_step.get(step),
            align_with_boundary_shape=align_with_boundary_shape,
            slope_sign=slope_sign,
            use_dual_slope=use_dual_slope,
            split_r=split_r_by_load_step.get(step),
        )

    maps = {step: data["avg"] for step, data in out.items()}
    color_scales = _compute_color_scales(
        maps_by_geom=maps,
        color_range_mode=color_range_mode,
        std_factor=std_factor,
        global_minmax=global_minmax,
    )

    n = len(out)
    ncols = 2
    nrows = max(1, int(np.ceil(n / ncols)))
    fig, axes = plt.subplots(nrows, ncols, figsize=(14, 5 * nrows), constrained_layout=True)
    axes = np.atleast_1d(axes).ravel()

    for i, (step, data) in enumerate(out.items()):
        ax = axes[i]
        has_mesh = data["r_mesh"] is not None and data["z_mesh"] is not None
        has_values = np.isfinite(data["avg"]).any()

        if has_mesh and has_values:
            mesh = ax.pcolormesh(
                data["r_mesh"],
                data["z_mesh"],
                data["avg"],
                shading="auto",
                cmap=cmap,
                vmin=color_scales[step][0],
                vmax=color_scales[step][1],
            )
            cbar = plt.colorbar(mesh, ax=ax, pad=0.01)
            cbar.set_label(property_name)
        else:
            ax.text(0.5, 0.5, "No valid data", ha="center", va="center", transform=ax.transAxes)

        mode_tag = "dual" if use_dual_slope else "single"
        ax.set_title(
            f"{step} | {property_name} x({property_sign:+d}) | angle={angle_deg:.1f}deg | k={data['incline_k']:.3f} | {mode_tag}",
            fontsize=10,
        )
        ax.set_xlabel("r")
        ax.set_ylabel("z (bottom-origin)")
        if invert_z_plot:
            ax.invert_yaxis()
            ax.set_ylabel("z (inverted for display)")

    for j in range(len(out), len(axes)):
        axes[j].axis("off")

    return fig, out, color_scales


# ------------------ Example usage ------------------
exp_graph_dict_file = Path(
    "/Users/yfjin/Desktop/PhD/CHESS02142024/GraphNetwork/AnalysisResults/Experiment/"
    "Config0_Sam1/graph_dict.pkl"
)

# For one experiment config, geometry angle is fixed across load steps.
# Set this to 0, 15, 30, 45... according to your sample config.
experiment_angle_deg = 0.0

# # Optional: step-specific split radius when dual slope is used.
# split_r_by_step = {
#     "load1": 3.75e-4,
#     "load2": 3.75e-4,
#     "load3": 3.75e-4,
# }

# Mode A: mean +/- n*std for each load step (default n=2).
fig_exp, exp_data, exp_color_scales = create_experiment_rz_binplots_from_graph_dict(
    graph_dict_path=exp_graph_dict_file,
    property_name="degree",  # this file does not contain node stress_hydro
    property_sign=1,
    graph_view="core",
    angle_deg=experiment_angle_deg,
    n_r_bins=10,
    n_z_bins=8,
    color_range_mode="mean_std_per_geometry",
    std_factor=2.0,
    align_with_boundary_shape=True,
    slope_sign=1,
    use_dual_slope=False,
    # split_r_by_load_step=split_r_by_step,
    invert_z_plot=True,
    cmap="RdBu_r",
)
plt.show()

print("Per-load-step color scales:")
for step, (vmin, vmax) in exp_color_scales.items():
    counts = exp_data[step]["counts"]
    print(f"  {step}: [{vmin:.4e}, {vmax:.4e}] | non-empty bins={np.count_nonzero(counts > 0)}")

# Mode B: shared custom min-max across all load steps.
# fig_exp_global, exp_data_global, exp_color_scales_global = create_experiment_rz_binplots_from_graph_dict(
#     graph_dict_path=exp_graph_dict_file,
#     property_name="avg_neighbor_degree",
#     property_sign=1,
#     graph_view="core",
#     angle_deg=experiment_angle_deg,
#     n_r_bins=13,
#     n_z_bins=10,
#     color_range_mode="custom_global_minmax",
#     global_minmax=(5.5, 8.0),
#     align_with_boundary_shape=True,
#     slope_sign=1,
#     use_dual_slope=False,
#     split_r_by_load_step=split_r_by_step,
#     invert_z_plot=True,
#     cmap="RdBu_r",
# )
# plt.show()
# print("Shared color scales:")
# for step, (vmin, vmax) in exp_color_scales_global.items():
#     print(f"  {step}: [{vmin:.4e}, {vmax:.4e}]")
```

```python
import re
import pickle
from collections import Counter
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from scipy.io import loadmat

EXP_GRAPH_DICT_PATH = Path(
    "/Users/yfjin/Desktop/PhD/CHESS02142024/GraphNetwork/AnalysisResults/Experiment/"
    "Config0_Sam1/graph_dict.pkl"
)
EXP_CONTACT_PATH = Path(
    "/Users/yfjin/Desktop/PhD/CHESS02142024/GraphNetwork/Experiment/Config0_Sam1/"
    "Tomo_Contacts/Config0_sam1_load2_ogcoords.mat"
)
EXP_STORED_PDF_DIR = Path(
    "/Users/yfjin/Desktop/PhD/Presentation/TMS2026/Granular_Figures/PlotData/0Degree"
)

EXP_GRAPH_VIEW = "core"  # core excludes wall contacts entirely
EXP_LOAD_STEP_INDEX = 1   # second load step after sorting: load2
EXP_LOAD_STEP_LABEL = None
COUNT_WALL_CONTACTS_ON_PARTICLES = False
WALL_ID = 1

BIN_CENTERS = np.arange(2, 12, 1)
BIN_EDGES = np.arange(1.5, 12.5, 1.0)

plt.rcParams["font.family"] = "Times New Roman"


def _exp_loadstep_sort_key(label):
    s = str(label)
    m = re.search(r"(\d+)", s)
    if m:
        return (0, int(m.group(1)), s)
    return (1, s.lower())


def _exp_is_graph_like(obj):
    return hasattr(obj, "nodes") and hasattr(obj, "edges")


def _exp_collect_graphs_from_entry(entry, graph_view="core"):
    if _exp_is_graph_like(entry):
        return [entry]

    if isinstance(entry, (list, tuple)):
        graphs = []
        for item in entry:
            graphs.extend(_exp_collect_graphs_from_entry(item, graph_view=graph_view))
        return graphs

    if isinstance(entry, dict):
        if graph_view in entry:
            preferred = _exp_collect_graphs_from_entry(entry[graph_view], graph_view=graph_view)
            if preferred:
                return preferred

        for key in ("graph", "graphs", "core", "full", "core_graph", "full_graph"):
            if key in entry:
                found = _exp_collect_graphs_from_entry(entry[key], graph_view=graph_view)
                if found:
                    return found

        graphs = []
        for value in entry.values():
            graphs.extend(_exp_collect_graphs_from_entry(value, graph_view=graph_view))
        return graphs

    return []


def _degree_pdf(values, bin_edges):
    values = np.asarray(values, dtype=float)
    if values.size == 0:
        raise ValueError("No degree values available for PDF calculation.")
    hist, _ = np.histogram(values, bins=bin_edges, density=True)
    return hist


def _load_stored_curve(path, key):
    arr = np.asarray(loadmat(path)[key], dtype=float).squeeze()
    if arr.size != BIN_CENTERS.size:
        raise ValueError(f"Stored curve '{key}' has {arr.size} points; expected {BIN_CENTERS.size}.")
    return arr


def _plot_filled_hist(ax, values, color, label, alpha=0.18, linewidth=2.0, linestyle="-"):
    ax.stairs(values, BIN_EDGES, fill=True, color=color, alpha=alpha, linewidth=0)
    ax.stairs(values, BIN_EDGES, fill=False, color=color, linewidth=linewidth, linestyle=linestyle, label=label)


def _contact_degrees_from_mat(contact_path, wall_id=1, count_wall_contacts_on_particles=True):
    og_contact = np.asarray(loadmat(contact_path)["og_contact"], dtype=float)
    raw_pairs = og_contact[:, :2].astype(int)
    unique_pairs = np.unique(np.sort(raw_pairs, axis=1), axis=0)

    degree_counter = Counter()
    if count_wall_contacts_on_particles:
        for n1, n2 in unique_pairs:
            if n1 != wall_id:
                degree_counter[int(n1)] += 1
            if n2 != wall_id:
                degree_counter[int(n2)] += 1
    else:
        non_wall_pairs = unique_pairs[(unique_pairs[:, 0] != wall_id) & (unique_pairs[:, 1] != wall_id)]
        for n1, n2 in non_wall_pairs:
            degree_counter[int(n1)] += 1
            degree_counter[int(n2)] += 1
        unique_pairs = non_wall_pairs

    degrees = np.asarray(list(degree_counter.values()), dtype=float)
    return degrees, raw_pairs, unique_pairs


with EXP_GRAPH_DICT_PATH.open("rb") as f:
    exp_graph_dict = pickle.load(f)

load_labels = sorted(exp_graph_dict.keys(), key=_exp_loadstep_sort_key)
selected_load = EXP_LOAD_STEP_LABEL or load_labels[EXP_LOAD_STEP_INDEX]
graphs = _exp_collect_graphs_from_entry(exp_graph_dict[selected_load], graph_view=EXP_GRAPH_VIEW)

if len(graphs) == 0:
    raise ValueError(f"No graphs found for {selected_load} in view '{EXP_GRAPH_VIEW}'.")

graph_degrees = []
graph_node_count = 0
graph_edge_count = 0
for G in graphs:
    graph_node_count += G.number_of_nodes()
    graph_edge_count += G.number_of_edges()
    for node, data in G.nodes(data=True):
        if data.get("is_wall", False):
            continue
        graph_degrees.append(float(G.degree(node)))

graph_degrees = np.asarray(graph_degrees, dtype=float)
contact_degrees, raw_contact_pairs, unique_contact_pairs = _contact_degrees_from_mat(
    EXP_CONTACT_PATH,
    wall_id=WALL_ID,
    count_wall_contacts_on_particles=COUNT_WALL_CONTACTS_ON_PARTICLES,
)

stored_curves = {
    "Stored min": _load_stored_curve(EXP_STORED_PDF_DIR / "Z_min_exp_0.mat", "Zmin"),
    "Stored med": _load_stored_curve(EXP_STORED_PDF_DIR / "Z_med_exp_0.mat", "Zmed"),
    "Stored max": _load_stored_curve(EXP_STORED_PDF_DIR / "Z_max_exp_0.mat", "Zmax"),
}

graph_pdf = _degree_pdf(graph_degrees, BIN_EDGES)
contact_pdf = _degree_pdf(contact_degrees, BIN_EDGES)
max_abs_pdf_diff = float(np.max(np.abs(graph_pdf - contact_pdf)))

fig, ax = plt.subplots(figsize=(8, 5.5))
_plot_filled_hist(ax, graph_pdf, color="#0b7bcb", label=f"Graph ({EXP_GRAPH_VIEW})", alpha=0.22, linewidth=2.4)
_plot_filled_hist(ax, contact_pdf, color="#ff7f0e", label="Contact list", alpha=0.18, linewidth=2.2, linestyle="--")
_plot_filled_hist(ax, stored_curves["Stored min"], color="#2ca02c", label="Stored min", alpha=0.12, linewidth=2.0, linestyle=":")
_plot_filled_hist(ax, stored_curves["Stored med"], color="#333333", label="Stored med", alpha=0.10, linewidth=2.0, linestyle="-.")
_plot_filled_hist(ax, stored_curves["Stored max"], color="#d62728", label="Stored max", alpha=0.12, linewidth=2.0, linestyle=":")

ax.set_xticks(BIN_CENTERS)
ax.set_xlabel("Degree", fontsize=18)
ax.set_ylabel("PDF", fontsize=18)
ax.set_title(f"Experimental Degree PDF Comparison: {selected_load} ({EXP_GRAPH_VIEW})", fontsize=18)
ax.tick_params(axis="both", which="major", labelsize=14, width=1, length=5)
for spine in ax.spines.values():
    spine.set_linewidth(1)
ax.legend(frameon=False, fontsize=12, ncol=2)
fig.tight_layout()
plt.show()

print(f"Selected load step: {selected_load}")
print(f"Graph view: {EXP_GRAPH_VIEW}")
print(f"Count wall contacts on particles: {COUNT_WALL_CONTACTS_ON_PARTICLES}")
print(f"Graph particle-node degrees used: {graph_degrees.size}")
print(f"Contact-list particle degrees used: {contact_degrees.size}")
print(f"Raw og_contact rows: {raw_contact_pairs.shape[0]}")
print(f"Unique unordered contact pairs used: {unique_contact_pairs.shape[0]}")
print(f"Graph total nodes (including wall nodes if present): {graph_node_count}")
print(f"Graph total edges: {graph_edge_count}")
print(f"Graph mean degree: {np.mean(graph_degrees):.6f}")
print(f"Contact-list mean degree: {np.mean(contact_degrees):.6f}")
print(f"Graph degree variance: {np.var(graph_degrees, ddof=0):.6f}")
print(f"Contact-list degree variance: {np.var(contact_degrees, ddof=0):.6f}")
print(f"Max |graph PDF - contact PDF| over bins 2:1:11: {max_abs_pdf_diff:.6e}")
```

```python
import random
import pickle
from pathlib import Path

import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# ── style: aligned with your previous HF visualization ───────────
EDGE_GRAY = mcolors.to_rgba("#bdbdbd")
NODE_GRAY = mcolors.to_rgba("#7f7f7f")

# requested style for "type" view
HF_TYPE_EDGE_COLOR = mcolors.to_rgba("#00274C")  # dark blue contacts
HF_TYPE_NODE_COLOR = mcolors.to_rgba("#C0392B")  # red connected nodes

EDGE_W_OTHER = 0.75
EDGE_W_HF = 1.5
ALPHA_OTHER = 0.5
ALPHA_HF = 0.90

NODE_SIZE_PART = 20
NODE_SIZE_WALL = 26
NODE_ALPHA_HF = 0.95
NODE_ALPHA_OTHER = 0.5


def _pos(G, n):
    """Particle: node['position']; Wall: avg of incident 'contact_location'."""
    if not G.nodes[n].get("is_wall", False):
        x, y, z = G.nodes[n]["position"]
        return float(x), float(y), float(z)

    locs = []
    if G.is_multigraph():
        for _, _, _, d in G.edges(n, keys=True, data=True):
            if "contact_location" in d:
                locs.append(d["contact_location"])
    else:
        for _, _, d in G.edges(n, data=True):
            if "contact_location" in d:
                locs.append(d["contact_location"])

    if locs:
        xs, ys, zs = zip(*locs)
        return (sum(xs) / len(xs), sum(ys) / len(ys), sum(zs) / len(zs))

    return (0.0, 0.0, 0.0)


def _iter_edges_with_data(graph):
    if graph.is_multigraph():
        for u, v, _, d in graph.edges(keys=True, data=True):
            yield u, v, d
    else:
        for u, v, d in graph.edges(data=True):
            yield u, v, d


def _equal_axes_3d(ax, xs, ys, zs):
    xr = max(xs) - min(xs)
    yr = max(ys) - min(ys)
    zr = max(zs) - min(zs)
    R = max(xr, yr, zr) or 1.0
    cx = 0.5 * (max(xs) + min(xs))
    cy = 0.5 * (max(ys) + min(ys))
    cz = 0.5 * (max(zs) + min(zs))

    ax.set_xlim(cx - R / 2, cx + R / 2)
    ax.set_ylim(cy - R / 2, cy + R / 2)
    ax.set_zlim(cz - R / 2, cz + R / 2)


def _cluster_color(cid):
    """Deterministic color per cluster id for non-(-100) clusters."""
    cmap = plt.get_cmap("tab20")
    return mcolors.to_rgba(cmap(((int(cid) - 1) % 20) / 19.0))


def _transparent_3d(ax, fig):
    """Remove visual frame so exported overlays stay clean."""
    fig.patch.set_alpha(0.0)
    ax.set_facecolor((1.0, 1.0, 1.0, 0.0))

    for axis in (ax.xaxis, ax.yaxis, ax.zaxis):
        axis.pane.set_facecolor((1.0, 1.0, 1.0, 0.0))
        axis.pane.set_edgecolor((1.0, 1.0, 1.0, 0.0))

    ax.set_axis_off()


def plot_hf_3d(G, color_mode="id", elev=25, azim=60, n_ticks=3, angle_vmin=None, angle_vmax=None):
    """
    Plot full graph in 3D.

    color_mode:
      - "id"    : HF edges colored by hf_cluster_id (type -100 gray)
      - "type"  : full graph + HF contacts dark blue + connected nodes red
      - "angle" : HF edges colored by angle_with_zz (viridis)

    angle_vmin / angle_vmax:
      Optional fixed colorbar limits for angle mode.
    """
    fig = plt.figure(figsize=(6.5, 6), facecolor=(1, 1, 1, 0))
    ax = fig.add_subplot(111, projection="3d")
    ax.view_init(elev=elev, azim=azim)

    xs, ys, zs = [], [], []
    for n in G.nodes:
        x, y, z = _pos(G, n)
        xs.append(x)
        ys.append(y)
        zs.append(z)

    # Draw non-HF edges in all modes so type keeps background structure too.
    for u, v, d in _iter_edges_with_data(G):
        if d.get("is_high_force", False):
            continue
        x1, y1, z1 = _pos(G, u)
        x2, y2, z2 = _pos(G, v)
        ax.plot([x1, x2], [y1, y2], [z1, z2], color=EDGE_GRAY, lw=EDGE_W_OTHER, alpha=ALPHA_OTHER)

    # Precompute angle color scale for HF edges if needed.
    angle_norm = None
    angle_cmap = plt.get_cmap("viridis")
    if color_mode == "angle":
        if (angle_vmin is not None) and (angle_vmax is not None):
            angle_norm = mcolors.Normalize(vmin=float(angle_vmin), vmax=float(angle_vmax))
        else:
            angles = []
            for _, _, d in _iter_edges_with_data(G):
                if d.get("is_high_force", False) and d.get("hf_cluster_type", 1) != -100:
                    ang = d.get("angle_with_zz", None)
                    if ang is not None:
                        angles.append(float(ang))
            if angles:
                amin, amax = min(angles), max(angles)
                if amin == amax:
                    amin, amax = amin - 1.0, amax + 1.0
                angle_norm = mcolors.Normalize(vmin=amin, vmax=amax)

    # Draw HF edges and assign node colors.
    node_color = {}

    def _hf_color(d):
        ctype = d.get("hf_cluster_type", 1)
        if ctype == -100:
            return EDGE_GRAY

        if color_mode == "id":
            return _cluster_color(d.get("hf_cluster_id", 1))

        if color_mode == "type":
            return HF_TYPE_EDGE_COLOR

        if color_mode == "angle":
            ang = d.get("angle_with_zz", None)
            if (ang is None) or (angle_norm is None):
                return EDGE_GRAY
            return mcolors.to_rgba(angle_cmap(angle_norm(float(ang))))

        raise ValueError("color_mode must be one of: id, type, angle")

    for u, v, d in _iter_edges_with_data(G):
        if not d.get("is_high_force", False):
            continue

        color = _hf_color(d)
        x1, y1, z1 = _pos(G, u)
        x2, y2, z2 = _pos(G, v)
        ax.plot([x1, x2], [y1, y2], [z1, z2], color=color, lw=EDGE_W_HF, alpha=ALPHA_HF)

        if color_mode == "type":
            node_color[u] = HF_TYPE_NODE_COLOR
            node_color[v] = HF_TYPE_NODE_COLOR
        else:
            if u not in node_color:
                node_color[u] = color
            if v not in node_color:
                node_color[v] = color

    def _scatter(nodes, marker, colors, alpha, size):
        if not nodes:
            return
        XYZ = [_pos(G, n) for n in nodes]
        X, Y, Z = zip(*XYZ)
        ax.scatter(X, Y, Z, marker=marker, s=size, c=colors, alpha=alpha, edgecolors="none")

    wall = [n for n in G.nodes if G.nodes[n].get("is_wall", False)]
    part = [n for n in G.nodes if not G.nodes[n].get("is_wall", False)]

    part_other = [n for n in part if n not in node_color]
    wall_other = [n for n in wall if n not in node_color]
    part_hf = [n for n in part if n in node_color]
    wall_hf = [n for n in wall if n in node_color]

    _scatter(part_other, "o", [NODE_GRAY] * len(part_other), NODE_ALPHA_OTHER, NODE_SIZE_PART)
    _scatter(wall_other, "^", [NODE_GRAY] * len(wall_other), NODE_ALPHA_OTHER, NODE_SIZE_WALL)
    _scatter(part_hf, "o", [node_color[n] for n in part_hf], NODE_ALPHA_HF, NODE_SIZE_PART)
    _scatter(wall_hf, "^", [node_color[n] for n in wall_hf], NODE_ALPHA_HF, NODE_SIZE_WALL)

    _equal_axes_3d(ax, xs, ys, zs)

    # Keep locators defined for consistent framing (hidden visually by set_axis_off).
    ax.xaxis.set_major_locator(plt.MaxNLocator(n_ticks))
    ax.yaxis.set_major_locator(plt.MaxNLocator(n_ticks))
    ax.zaxis.set_major_locator(plt.MaxNLocator(n_ticks))

    if (color_mode == "angle") and (angle_norm is not None):
        sm = cm.ScalarMappable(norm=angle_norm, cmap=angle_cmap)
        sm.set_array([])
        cbar = fig.colorbar(sm, ax=ax, fraction=0.035, pad=0.02)
        cbar.set_label("angle_with_zz")

    _transparent_3d(ax, fig)
    plt.tight_layout(pad=0)
    return fig, ax


def sample_graph_indices_per_geometry(GL_dict, n_per_geom=4, seed=0):
    rng = random.Random(seed)
    picks = {}
    for geom, dd in GL_dict.items():
        n = len(dd["full"])
        if n == 0:
            picks[geom] = []
            continue
        k = min(n_per_geom, n)
        picks[geom] = sorted(rng.sample(range(n), k))
    return picks


def get_global_angle_range(GL_dict, indices_by_geom):
    """Collect one shared angle range across all selected graphs for batch comparability."""
    vals = []
    for geom in sorted(GL_dict.keys()):
        graphs = GL_dict[geom]["full"]
        for sim_idx in indices_by_geom.get(geom, []):
            G = graphs[sim_idx]
            for _, _, d in _iter_edges_with_data(G):
                if not d.get("is_high_force", False):
                    continue
                if d.get("hf_cluster_type", 1) == -100:
                    continue
                ang = d.get("angle_with_zz", None)
                if ang is not None:
                    vals.append(float(ang))

    if not vals:
        return None, None

    vmin, vmax = min(vals), max(vals)
    if vmin == vmax:
        vmin, vmax = vmin - 1.0, vmax + 1.0
    return vmin, vmax


def batch_plot_hf_modes(
    GL_dict,
    indices_by_geom,
    color_modes=("type", "angle"),
    elev=20,
    azim=30,
    show=False,
    save=True,
    dpi=170,
    out_dir=None,
):
    if out_dir is None:
        out_dir = PLOTS_DIR / "HF_Clusters"
    else:
        out_dir = Path(out_dir)

    if save:
        out_dir.mkdir(parents=True, exist_ok=True)

    angle_vmin, angle_vmax = None, None
    if "angle" in color_modes:
        angle_vmin, angle_vmax = get_global_angle_range(GL_dict, indices_by_geom)
        print(f"Global angle_with_zz range for batch colorbar: [{angle_vmin}, {angle_vmax}]")

    for geom in sorted(GL_dict.keys()):
        graphs = GL_dict[geom]["full"]
        for sim_idx in indices_by_geom.get(geom, []):
            G = graphs[sim_idx]
            for mode in color_modes:
                fig, _ = plot_hf_3d(
                    G,
                    color_mode=mode,
                    elev=elev,
                    azim=azim,
                    angle_vmin=angle_vmin if mode == "angle" else None,
                    angle_vmax=angle_vmax if mode == "angle" else None,
                )

                if save:
                    fname = f"hf_{mode}_{geom}_sim{sim_idx:02d}_e{elev}_a{azim}.png"
                    fpath = out_dir / fname
                    fig.savefig(fpath, dpi=dpi, bbox_inches="tight", transparent=True)
                    print(f"Saved: {fpath}")

                if show:
                    plt.show()
                else:
                    plt.close(fig)


# ── load labeled simulation graph dictionary ──────────────────────
pkl_path = GRAPH_WITH_CLUSTERS_PATH

with open(pkl_path, "rb") as f:
    GL = pickle.load(f)

# ── pick 4 simulations per geometry (random but reproducible) ─────
indices_by_geom = sample_graph_indices_per_geometry(GL, n_per_geom=4, seed=7)
print("Selected simulation indices:", indices_by_geom)

# ── quick preview on one graph per mode (fast sanity check) ───────
preview_geom = "45deg"
preview_idx = indices_by_geom[preview_geom][0]
preview_G = GL[preview_geom]["full"][preview_idx]

preview_vmin, preview_vmax = get_global_angle_range(GL, indices_by_geom)
for mode in ("id", "type", "angle"):
    fig, _ = plot_hf_3d(
        preview_G,
        color_mode=mode,
        elev=20,
        azim=30,
        angle_vmin=preview_vmin if mode == "angle" else None,
        angle_vmax=preview_vmax if mode == "angle" else None,
    )
    plt.show()
    plt.close(fig)

# ── run batch only when needed (can be slow) ──────────────────────
RUN_BATCH = True
if RUN_BATCH:
    batch_plot_hf_modes(
        GL_dict=GL,
        indices_by_geom=indices_by_geom,
        # color_modes=("type", "angle", 'id'),
        color_modes=("type", "angle"),
        elev=20,
        azim=30,
        show=False,
        save=True,
        dpi=170,
    )
    print("Done: HF cluster views saved (type/angle).")
```

### 4.5.1) Curvature vs. normal force bin plots

```python
# 5 plots: 0deg, 15deg, 30deg, 45deg, and ALL (combined)
# Binned 2D density of curvature_no_walls vs normal_force with shared color range.

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

# -----------------------------
# Configuration
# -----------------------------
GRAPH_VIEW = 'core'                 # 'core' or 'full'
X_ATTR = 'curvature_no_walls'       # x-axis attribute
Y_ATTR = 'normal_force'             # y-axis attribute
BINS_X = 40
BINS_Y = 40
CMAP = 'coolwarm'

# Optional manual axis limits (set to None for auto from data)
X_RANGE = None
Y_RANGE = None

# Plot style
FIGSIZE = (7, 5.5)
AXIS_LABEL_SIZE = 14
TICK_LABEL_SIZE = 12

# -----------------------------
# Data extraction
# -----------------------------
def collect_edge_attr_pairs(graph_dict, geom, graph_view='core', x_attr='curvature_no_walls', y_attr='normal_force'):
    """Collect valid (x_attr, y_attr) pairs from all simulations of one geometry."""
    xs, ys = [], []
    graphs = graph_dict[geom][graph_view]

    for G in graphs:
        for _, _, d in G.edges(data=True):
            xv = d.get(x_attr, np.nan)
            yv = d.get(y_attr, np.nan)
            try:
                xv = float(xv)
                yv = float(yv)
            except (TypeError, ValueError):
                continue

            if np.isfinite(xv) and np.isfinite(yv):
                xs.append(xv)
                ys.append(yv)

    return np.asarray(xs), np.asarray(ys)


geoms = ['0deg', '15deg', '30deg', '45deg']
per_geom = {}

for g in geoms:
    if g not in GL_final:
        print(f"Warning: geometry {g} not found in GL_final; skipping")
        continue
    xg, yg = collect_edge_attr_pairs(GL_final, g, graph_view=GRAPH_VIEW, x_attr=X_ATTR, y_attr=Y_ATTR)
    per_geom[g] = (xg, yg)

# Combine all geometries
x_all = np.concatenate([per_geom[g][0] for g in per_geom]) if per_geom else np.array([])
y_all = np.concatenate([per_geom[g][1] for g in per_geom]) if per_geom else np.array([])

if x_all.size == 0:
    raise ValueError('No valid edge attribute pairs found for requested attributes/graph view.')

# Shared x/y ranges across all figures
if X_RANGE is None:
    x_min, x_max = np.nanmin(x_all), np.nanmax(x_all)
else:
    x_min, x_max = X_RANGE

if Y_RANGE is None:
    y_min, y_max = np.nanmin(y_all), np.nanmax(y_all)
else:
    y_min, y_max = Y_RANGE

if x_min == x_max:
    x_min -= 1e-9
    x_max += 1e-9
if y_min == y_max:
    y_min -= 1e-9
    y_max += 1e-9

x_edges = np.linspace(x_min, x_max, BINS_X + 1)
y_edges = np.linspace(y_min, y_max, BINS_Y + 1)

# Build histograms first so all plots share the same color range
histograms = {}
all_names = geoms + ['all']

for name in all_names:
    if name == 'all':
        xvals, yvals = x_all, y_all
    else:
        if name not in per_geom:
            continue
        xvals, yvals = per_geom[name]

    H, _, _ = np.histogram2d(xvals, yvals, bins=[x_edges, y_edges])
    histograms[name] = H

# Log scale needs strictly positive vmin
all_positive_counts = np.concatenate([H[H > 0].ravel() for H in histograms.values() if np.any(H > 0)])
if all_positive_counts.size == 0:
    raise ValueError('All bins are zero; cannot use log color scale.')

vmin = float(np.min(all_positive_counts))
vmax = float(np.max(all_positive_counts))
if vmax <= vmin:
    vmax = vmin * 10.0

norm = LogNorm(vmin=vmin, vmax=vmax)

# -----------------------------
# Plot 5 separate binned figures
# -----------------------------
for name in all_names:
    if name not in histograms:
        continue

    H = histograms[name]
    H_masked = np.ma.masked_less_equal(H.T, 0)

    fig, ax = plt.subplots(figsize=FIGSIZE)
    mesh = ax.pcolormesh(x_edges, y_edges, H_masked, cmap=CMAP, norm=norm, shading='auto')

    cbar = fig.colorbar(mesh, ax=ax)
    cbar.set_label('Edge count per bin (log scale)', fontsize=AXIS_LABEL_SIZE)
    cbar.ax.tick_params(labelsize=TICK_LABEL_SIZE)

    title = f"{name}: {X_ATTR} vs {Y_ATTR} (binned, log color)"
    ax.set_title(title, fontsize=AXIS_LABEL_SIZE)
    ax.set_xlabel(X_ATTR, fontsize=AXIS_LABEL_SIZE)
    ax.set_ylabel(Y_ATTR, fontsize=AXIS_LABEL_SIZE)
    ax.tick_params(axis='both', which='major', labelsize=TICK_LABEL_SIZE)

    fig.tight_layout()
    plt.show()

print('Done: generated 5 separate binned plots with shared LOG color range (coolwarm).')

# -----------------------------
# Linear fit right after bin plots
# -----------------------------
def linear_fit_metrics(x, y):
    """Return slope, intercept, and R^2 for y = m*x + b."""
    m, b = np.polyfit(x, y, 1)
    yhat = m * x + b
    ss_res = float(np.sum((y - yhat) ** 2))
    ss_tot = float(np.sum((y - np.mean(y)) ** 2))
    r2 = 1.0 - ss_res / ss_tot if ss_tot > 0 else np.nan
    return float(m), float(b), float(r2)

rng = np.random.default_rng(0)
MAX_SCATTER_POINTS = 20000
fit_stats = {}

for name in all_names:
    if name == 'all':
        xvals, yvals = x_all, y_all
    else:
        if name not in per_geom:
            continue
        xvals, yvals = per_geom[name]

    if len(xvals) < 2:
        print(f"Skipping fit for {name}: not enough points")
        continue

    m, b, r2 = linear_fit_metrics(xvals, yvals)
    fit_stats[name] = (m, b, r2, len(xvals))

    # Downsample scatter for rendering speed
    if len(xvals) > MAX_SCATTER_POINTS:
        idx = rng.choice(len(xvals), size=MAX_SCATTER_POINTS, replace=False)
        xs, ys = xvals[idx], yvals[idx]
    else:
        xs, ys = xvals, yvals

    fig, ax = plt.subplots(figsize=FIGSIZE)
    ax.scatter(xs, ys, s=4, alpha=0.15, color='#4c72b0', edgecolors='none')

    xx = np.linspace(x_min, x_max, 200)
    yy = m * xx + b
    ax.plot(xx, yy, color='#d62728', linewidth=2.5,
            label=f"fit: y = {m:.4g}x + {b:.4g}\n$R^2$ = {r2:.4f}")

    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)
    ax.set_title(f"{name}: linear fit of {Y_ATTR} vs {X_ATTR}", fontsize=AXIS_LABEL_SIZE)
    ax.set_xlabel(X_ATTR, fontsize=AXIS_LABEL_SIZE)
    ax.set_ylabel(Y_ATTR, fontsize=AXIS_LABEL_SIZE)
    ax.tick_params(axis='both', which='major', labelsize=TICK_LABEL_SIZE)
    ax.legend(frameon=False, fontsize=10)

    fig.tight_layout()
    plt.show()

print('\nLinear fit summary (y = m*x + b):')
for name in all_names:
    if name in fit_stats:
        m, b, r2, n = fit_stats[name]
        print(f"{name:>5s} | n={n:7d} | m={m: .6e} | b={b: .6e} | R^2={r2:.5f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from pathlib import Path
import pickle

plt.rcParams["font.family"] = "Times New Roman"

SEC6_GRAPH_VIEW = "core"
SEC6_GEOMS = ["0deg", "15deg", "30deg", "45deg"]
SEC6_GEOM_TO_DEG = {"0deg": 0.0, "15deg": 15.0, "30deg": 30.0, "45deg": 45.0}

# Trim these lists if you want fewer figures.
SEC6_NODE_PROPERTIES = [
    "degree",
    "clustering",
    "avg_neighbor_degree",
    "betweenness",
    "closeness",
    "fiedler",
    "principal_eigenvector",
    "avg_curvature_no_walls",
    "high_force_degree",
    "nfd",
    "nfd_r2",
]

SEC6_EDGE_PROPERTIES = [
    "curvature_no_walls",
    "edge_connectivity",
    "node_connectivity",
    "angle_with_zz",
]

SEC6_NODE_CORR_FEATURES = [
    name for name in SEC6_NODE_PROPERTIES if name != "high_force_degree"
] + ["stress_33"]

SEC6_EDGE_CORR_FEATURES = [
    "curvature_no_walls",
    "node_connectivity",
    "angle_with_zz",
]

SEC6_NODE_TARGET_ATTR = "stress_hydro"
SEC6_NODE_TARGET_NAME = "hydrostress_mag"
SEC6_NODE_TARGET_SIGN = -1.0  # plot compressive hydrostress as positive magnitude
SEC6_EDGE_TARGET_ATTR = "normal_force"
SEC6_EDGE_TARGET_NAME = "normal_force"

SEC6_NODE_BINS_X = 40
SEC6_NODE_BINS_Y = 40
SEC6_EDGE_BINS_X = 40
SEC6_EDGE_BINS_Y = 40
SEC6_BIN_CMAP = "coolwarm"
SEC6_NODE_X_RANGES = {}
SEC6_EDGE_X_RANGES = {}
SEC6_NODE_Y_RANGE = None
SEC6_EDGE_Y_RANGE = None

SEC6_FIGSIZE = (24, 4.8)
SEC6_AXIS_LABEL_SIZE = 12
SEC6_TITLE_SIZE = 13
SEC6_TICK_LABEL_SIZE = 10
SEC6_PCA_MAX_POINTS = 20000
SEC6_RNG = np.random.default_rng(6)

SEC6_ATTR_LABELS = {
    "degree": "Degree",
    "clustering": "Clustering coefficient",
    "avg_neighbor_degree": "Average neighbor degree",
    "betweenness": "Betweenness",
    "closeness": "Closeness",
    "fiedler": "Fiedler value",
    "principal_eigenvector": "Principal eigenvector",
    "avg_curvature_no_walls": "Average curvature (no walls)",
    "high_force_degree": "High-force degree",
    "nfd": "NFD",
    "nfd_r2": "NFD $R^2$",
    "curvature_no_walls": "Curvature (no walls)",
    "edge_connectivity": "Edge connectivity",
    "node_connectivity": "Node connectivity",
    "delta": "Normal overlap delta",
    "delta_t": "Tangential overlap delta_t",
    "angle_with_zz": "Angle with zz (deg)",
    "tangential_force": "Tangential force",
    "stress_33": "Stress 33 (loading direction)",
    "hydrostress_mag": "Compressive hydrostress (-stress_hydro)",
    "normal_force": "Normal force",
    "loading_deg": "Loading angle (deg)",
}


def _sec6_label(name):
    return SEC6_ATTR_LABELS.get(name, name)


def _sec6_is_graph_like(obj):
    return hasattr(obj, "nodes") and hasattr(obj, "edges")


def _sec6_collect_graphs(entry, graph_view="core"):
    if _sec6_is_graph_like(entry):
        return [entry]

    if isinstance(entry, (list, tuple)):
        graphs = []
        for item in entry:
            graphs.extend(_sec6_collect_graphs(item, graph_view=graph_view))
        return graphs

    if isinstance(entry, dict):
        if graph_view in entry:
            graphs = _sec6_collect_graphs(entry[graph_view], graph_view=graph_view)
            if graphs:
                return graphs

        graphs = []
        for value in entry.values():
            graphs.extend(_sec6_collect_graphs(value, graph_view=graph_view))
        return graphs

    return []


def _sec6_safe_float(value):
    try:
        return float(value)
    except (TypeError, ValueError):
        return np.nan


def _sec6_load_graph_dict():
    if "GL_final" in globals():
        return GL_final
    if "GL_prop" in globals():
        return GL_prop

    path = GRAPH_LABELED_PATH
    with path.open("rb") as f:
        return pickle.load(f)


SEC6_GL = _sec6_load_graph_dict()


def _sec6_iter_graphs(graph_dict, geom, graph_view="core"):
    if geom not in graph_dict:
        return []
    return _sec6_collect_graphs(graph_dict[geom], graph_view=graph_view)


def _sec6_collect_pairs(
    graph_dict,
    geoms,
    *,
    element,
    x_attr,
    y_attr,
    y_sign=1.0,
    graph_view="core",
):
    per_geom = {}

    for geom in geoms:
        xs, ys = [], []
        for G in _sec6_iter_graphs(graph_dict, geom, graph_view=graph_view):
            if element == "node":
                for _, data in G.nodes(data=True):
                    xv = _sec6_safe_float(data.get(x_attr))
                    yv = y_sign * _sec6_safe_float(data.get(y_attr))
                    if np.isfinite(xv) and np.isfinite(yv):
                        xs.append(xv)
                        ys.append(yv)
            elif element == "edge":
                for _, _, data in G.edges(data=True):
                    xv = _sec6_safe_float(data.get(x_attr))
                    yv = y_sign * _sec6_safe_float(data.get(y_attr))
                    if np.isfinite(xv) and np.isfinite(yv):
                        xs.append(xv)
                        ys.append(yv)
            else:
                raise ValueError("element must be 'node' or 'edge'")

        if xs:
            per_geom[geom] = (np.asarray(xs, dtype=float), np.asarray(ys, dtype=float))

    return per_geom


def _sec6_plot_binned_panels(
    per_geom,
    *,
    x_attr,
    y_name,
    bins_x,
    bins_y,
    cmap,
    count_label,
    x_range=None,
    y_range=None,
):
    panel_names = [geom for geom in SEC6_GEOMS if geom in per_geom]
    if not panel_names:
        raise ValueError(f"No valid data found for {x_attr}.")

    x_all = np.concatenate([per_geom[g][0] for g in panel_names])
    y_all = np.concatenate([per_geom[g][1] for g in panel_names])

    if x_range is None:
        x_min, x_max = float(np.nanmin(x_all)), float(np.nanmax(x_all))
    else:
        x_min, x_max = x_range

    if y_range is None:
        y_min, y_max = float(np.nanmin(y_all)), float(np.nanmax(y_all))
    else:
        y_min, y_max = y_range

    if x_min == x_max:
        x_min -= 1e-9
        x_max += 1e-9
    if y_min == y_max:
        y_min -= 1e-9
        y_max += 1e-9

    x_edges = np.linspace(x_min, x_max, bins_x + 1)
    y_edges = np.linspace(y_min, y_max, bins_y + 1)

    histograms = {}
    for name in panel_names + ["all"]:
        if name == "all":
            xvals, yvals = x_all, y_all
        else:
            xvals, yvals = per_geom[name]
        H, _, _ = np.histogram2d(xvals, yvals, bins=[x_edges, y_edges])
        histograms[name] = H

    positive = [H[H > 0].ravel() for H in histograms.values() if np.any(H > 0)]
    if not positive:
        raise ValueError(f"All bins are zero for {x_attr}.")

    all_positive_counts = np.concatenate(positive)
    vmin = float(np.min(all_positive_counts))
    vmax = float(np.max(all_positive_counts))
    if vmax <= vmin:
        vmax = vmin * 10.0
    norm = LogNorm(vmin=vmin, vmax=vmax)

    fig, axes = plt.subplots(
        1,
        len(panel_names) + 1,
        figsize=SEC6_FIGSIZE,
        sharex=True,
        sharey=True,
        constrained_layout=True,
    )
    axes = np.atleast_1d(axes)
    mesh = None

    for ax, name in zip(axes, panel_names + ["all"]):
        H = histograms[name]
        H_masked = np.ma.masked_less_equal(H.T, 0)
        mesh = ax.pcolormesh(x_edges, y_edges, H_masked, cmap=cmap, norm=norm, shading="auto")
        ax.set_title("All geometries" if name == "all" else name, fontsize=SEC6_TITLE_SIZE)
        ax.set_xlabel(_sec6_label(x_attr), fontsize=SEC6_AXIS_LABEL_SIZE)
        ax.tick_params(axis="both", which="major", labelsize=SEC6_TICK_LABEL_SIZE)

    axes[0].set_ylabel(_sec6_label(y_name), fontsize=SEC6_AXIS_LABEL_SIZE)

    cbar = fig.colorbar(mesh, ax=axes, shrink=0.92, pad=0.015)
    cbar.set_label(count_label, fontsize=SEC6_AXIS_LABEL_SIZE)
    cbar.ax.tick_params(labelsize=SEC6_TICK_LABEL_SIZE)

    fig.suptitle(
        f"{_sec6_label(x_attr)} vs {_sec6_label(y_name)} ({SEC6_GRAPH_VIEW})",
        fontsize=SEC6_TITLE_SIZE + 1,
    )
    plt.show()


def _sec6_build_feature_matrix(
    graph_dict,
    geoms,
    *,
    element,
    feature_names,
    target_attr=None,
    target_name=None,
    target_sign=1.0,
    include_loading=False,
    graph_view="core",
):
    rows = []
    geom_names = []

    for geom in geoms:
        for G in _sec6_iter_graphs(graph_dict, geom, graph_view=graph_view):
            if element == "node":
                data_items = (data for _, data in G.nodes(data=True))
            elif element == "edge":
                data_items = (data for _, _, data in G.edges(data=True))
            else:
                raise ValueError("element must be 'node' or 'edge'")

            for data in data_items:
                row = []
                valid = True
                for name in feature_names:
                    value = _sec6_safe_float(data.get(name))
                    if not np.isfinite(value):
                        valid = False
                        break
                    row.append(value)
                if not valid:
                    continue

                if target_attr is not None:
                    target_value = target_sign * _sec6_safe_float(data.get(target_attr))
                    if not np.isfinite(target_value):
                        continue
                    row.append(target_value)

                if include_loading:
                    row.append(float(SEC6_GEOM_TO_DEG[geom]))

                rows.append(row)
                geom_names.append(geom)

    column_names = list(feature_names)
    if target_attr is not None:
        column_names.append(target_name or target_attr)
    if include_loading:
        column_names.append("loading_deg")

    if not rows:
        raise ValueError("No valid rows found for section 6 analysis.")

    return {
        "matrix": np.asarray(rows, dtype=float),
        "column_names": column_names,
        "geom_names": np.asarray(geom_names),
    }


def _sec6_plot_correlation_matrix(matrix, column_names, *, title, figsize):
    corr = np.corrcoef(matrix, rowvar=False)
    corr = np.nan_to_num(corr, nan=0.0)

    fig, ax = plt.subplots(figsize=figsize)
    im = ax.imshow(corr, vmin=-1.0, vmax=1.0, cmap="coolwarm", aspect="auto")

    labels = [_sec6_label(name) for name in column_names]
    ax.set_xticks(np.arange(len(labels)))
    ax.set_yticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=45, ha="right", fontsize=SEC6_TICK_LABEL_SIZE)
    ax.set_yticklabels(labels, fontsize=SEC6_TICK_LABEL_SIZE)
    ax.set_title(title, fontsize=SEC6_TITLE_SIZE + 1)

    cbar = fig.colorbar(im, ax=ax, shrink=0.88)
    cbar.set_label("Pearson correlation", fontsize=SEC6_AXIS_LABEL_SIZE)
    cbar.ax.tick_params(labelsize=SEC6_TICK_LABEL_SIZE)

    fig.tight_layout()
    plt.show()
    return corr


def _sec6_print_top_correlations(corr, column_names, *, title, top_n=10):
    pairs = []
    for i in range(len(column_names)):
        for j in range(i + 1, len(column_names)):
            pairs.append((abs(corr[i, j]), corr[i, j], column_names[i], column_names[j]))

    pairs.sort(key=lambda item: item[0], reverse=True)
    print(f"\nTop {min(top_n, len(pairs))} absolute correlations: {title}")
    for _, corr_val, name_i, name_j in pairs[:top_n]:
        print(f"  {_sec6_label(name_i)} vs {_sec6_label(name_j)}: {corr_val:+.4f}")


def _sec6_run_pca(matrix, column_names):
    X = np.asarray(matrix, dtype=float)
    mean = np.mean(X, axis=0)
    std = np.std(X, axis=0, ddof=1)
    std[~np.isfinite(std) | (std == 0)] = 1.0

    X_std = (X - mean) / std
    U, S, Vt = np.linalg.svd(X_std, full_matrices=False)

    denom = max(X_std.shape[0] - 1, 1)
    explained_variance = (S ** 2) / denom
    explained_ratio = explained_variance / np.sum(explained_variance)
    scores = U * S

    return {
        "scores": scores,
        "components": Vt,
        "explained_ratio": explained_ratio,
        "feature_names": list(column_names),
    }


def _sec6_downsample_indices(n_points, max_points=None):
    if max_points is None or n_points <= max_points:
        return np.arange(n_points)
    return np.sort(SEC6_RNG.choice(n_points, size=max_points, replace=False))


def _sec6_plot_pca_overview(pca_result, color_values, *, color_name, title):
    scores = pca_result["scores"]
    explained_ratio = pca_result["explained_ratio"]
    feature_names = pca_result["feature_names"]
    components = pca_result["components"]

    idx = _sec6_downsample_indices(scores.shape[0], SEC6_PCA_MAX_POINTS)
    pc_count = min(6, len(explained_ratio))
    loading_count = min(3, components.shape[0])
    loading_matrix = components[:loading_count, :].T
    loading_scale = np.max(np.abs(loading_matrix))
    if loading_scale == 0:
        loading_scale = 1.0

    fig, axes = plt.subplots(1, 3, figsize=(18, 4.8))

    axes[0].plot(np.arange(1, pc_count + 1), explained_ratio[:pc_count] * 100.0, marker="o")
    axes[0].set_xlabel("Principal component", fontsize=SEC6_AXIS_LABEL_SIZE)
    axes[0].set_ylabel("Explained variance (%)", fontsize=SEC6_AXIS_LABEL_SIZE)
    axes[0].set_title("Scree plot", fontsize=SEC6_TITLE_SIZE)
    axes[0].tick_params(axis="both", which="major", labelsize=SEC6_TICK_LABEL_SIZE)

    im = axes[1].imshow(
        loading_matrix,
        aspect="auto",
        cmap="coolwarm",
        vmin=-loading_scale,
        vmax=loading_scale,
    )
    axes[1].set_xticks(np.arange(loading_count))
    axes[1].set_xticklabels([f"PC{i + 1}" for i in range(loading_count)], fontsize=SEC6_TICK_LABEL_SIZE)
    axes[1].set_yticks(np.arange(len(feature_names)))
    axes[1].set_yticklabels([_sec6_label(name) for name in feature_names], fontsize=SEC6_TICK_LABEL_SIZE)
    axes[1].set_title("Standardized feature loadings", fontsize=SEC6_TITLE_SIZE)
    cbar = fig.colorbar(im, ax=axes[1], shrink=0.88)
    cbar.ax.tick_params(labelsize=SEC6_TICK_LABEL_SIZE)

    scatter = axes[2].scatter(
        scores[idx, 0],
        scores[idx, 1],
        c=np.asarray(color_values)[idx],
        s=7,
        alpha=0.35,
        cmap="viridis",
        edgecolors="none",
    )
    axes[2].set_xlabel(f"PC1 ({explained_ratio[0] * 100.0:.1f}%)", fontsize=SEC6_AXIS_LABEL_SIZE)
    axes[2].set_ylabel(f"PC2 ({explained_ratio[1] * 100.0:.1f}%)", fontsize=SEC6_AXIS_LABEL_SIZE)
    axes[2].set_title(f"PC1 vs PC2, colored by {_sec6_label(color_name)}", fontsize=SEC6_TITLE_SIZE)
    axes[2].tick_params(axis="both", which="major", labelsize=SEC6_TICK_LABEL_SIZE)
    cbar = fig.colorbar(scatter, ax=axes[2], shrink=0.88)
    cbar.set_label(_sec6_label(color_name), fontsize=SEC6_AXIS_LABEL_SIZE)
    cbar.ax.tick_params(labelsize=SEC6_TICK_LABEL_SIZE)

    fig.suptitle(title, fontsize=SEC6_TITLE_SIZE + 1)
    fig.tight_layout(rect=[0, 0, 1, 0.94])
    plt.show()


def _sec6_print_top_loadings(pca_result, *, title, n_components=2, top_n=5):
    components = pca_result["components"]
    feature_names = pca_result["feature_names"]

    print(f"\nTop PCA loadings: {title}")
    for pc_idx in range(min(n_components, components.shape[0])):
        weights = components[pc_idx]
        order = np.argsort(np.abs(weights))[::-1][:top_n]
        print(f"  PC{pc_idx + 1}:")
        for feature_idx in order:
            print(f"    {_sec6_label(feature_names[feature_idx])}: {weights[feature_idx]:+.4f}")
```

```python
import numpy as np

def _drop_columns_for_pca(matrix, column_names, drop_names):
    """
    Return a PCA-ready matrix with selected columns removed, while also
    returning the dropped columns for later coloring/analysis.
    """
    col_to_idx = {name: i for i, name in enumerate(column_names)}

    dropped = {}
    for name in drop_names:
        if name in col_to_idx:
            dropped[name] = matrix[:, col_to_idx[name]].copy()

    keep_idx = [i for i, name in enumerate(column_names) if name not in drop_names]
    kept_matrix = matrix[:, keep_idx]
    kept_names = [column_names[i] for i in keep_idx]

    return {
        "matrix": kept_matrix,
        "column_names": kept_names,
        "dropped": dropped,
        "keep_idx": keep_idx,
    }

# ----------------------------
# Nodes: build full matrix first
# ----------------------------
node_full_data = _sec6_build_feature_matrix(
    SEC6_GL,
    SEC6_GEOMS,
    element="node",
    feature_names=SEC6_NODE_PROPERTIES,
    target_attr=SEC6_NODE_TARGET_ATTR,
    target_name=SEC6_NODE_TARGET_NAME,
    target_sign=SEC6_NODE_TARGET_SIGN,
    include_loading=True,
    graph_view=SEC6_GRAPH_VIEW,
)

# Save color variables from full matrix
node_hydro_idx = node_full_data["column_names"].index(SEC6_NODE_TARGET_NAME)
node_loading_idx = node_full_data["column_names"].index("loading_deg")

node_hydro = node_full_data["matrix"][:, node_hydro_idx]
node_loading = node_full_data["matrix"][:, node_loading_idx]

# Drop target + loading before PCA
node_pca_input = _drop_columns_for_pca(
    node_full_data["matrix"],
    node_full_data["column_names"],
    drop_names=[SEC6_NODE_TARGET_NAME, "loading_deg"],
)

node_pca = _sec6_run_pca(
    node_pca_input["matrix"],
    node_pca_input["column_names"],
)

# Plot PCA scores, colored afterward by external variables
_sec6_plot_pca_overview(
    node_pca,
    node_hydro,
    color_name=SEC6_NODE_TARGET_NAME,
    title="Node PCA: graph features only, colored by hydrostress",
)

_sec6_plot_pca_overview(
    node_pca,
    node_loading,
    color_name="loading_deg",
    title="Node PCA: graph features only, colored by loading angle",
)

print("Node explained variance (first 6 PCs):")
for pc_idx, ratio in enumerate(node_pca["explained_ratio"][:6], start=1):
    print(f"  PC{pc_idx}: {ratio * 100.0:6.2f}%")

_sec6_print_top_loadings(node_pca, title="nodes", n_components=2, top_n=5)
    
# ----------------------------
# Edges: build full matrix first
# ----------------------------
edge_full_data = _sec6_build_feature_matrix(
    SEC6_GL,
    SEC6_GEOMS,
    element="edge",
    feature_names=SEC6_EDGE_PROPERTIES,
    target_attr=SEC6_EDGE_TARGET_ATTR,
    target_name=SEC6_EDGE_TARGET_NAME,
    target_sign=1.0,
    include_loading=False,
    graph_view=SEC6_GRAPH_VIEW,
)

# Save force for coloring
edge_force_idx = edge_full_data["column_names"].index(SEC6_EDGE_TARGET_NAME)
edge_force = edge_full_data["matrix"][:, edge_force_idx]

# Drop target before PCA
edge_pca_input = _drop_columns_for_pca(
    edge_full_data["matrix"],
    edge_full_data["column_names"],
    drop_names=[SEC6_EDGE_TARGET_NAME],
)

edge_pca = _sec6_run_pca(
    edge_pca_input["matrix"],
    edge_pca_input["column_names"],
)

_sec6_plot_pca_overview(
    edge_pca,
    edge_force,
    color_name=SEC6_EDGE_TARGET_NAME,
    title="Edge PCA: graph features only, colored by normal force",
)

print("\nEdge explained variance (first 6 PCs):")
for pc_idx, ratio in enumerate(edge_pca["explained_ratio"][:6], start=1):
    print(f"  PC{pc_idx}: {ratio * 100.0:6.2f}%")

_sec6_print_top_loadings(edge_pca, title="edges", n_components=2, top_n=5)

```

```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
from pathlib import Path

# -----------------------------
# Config
# -----------------------------
GRAPH_VIEW = "core"      # "core" or "full"
USE_ALL_SIMS = True
SIM_INDEX = 0            # used only if USE_ALL_SIMS=False
GEOMS = ["0deg", "45deg"]

# Degree bins 2:1:11
# Bin centers: 2..11, edges: 1.5..11.5
BIN_EDGES = np.arange(0.5, 13.0, 1.0)
XTICKS = np.arange(1, 13, 1)

# Histogram mode:
# "percentage" -> sum of bars = 100%
# "pdf" -> probability density (MATLAB-style pdf normalization)
HIST_MODE = "percentage"

# Plot style
plt.rcParams["font.family"] = "Times New Roman"
figsize = (7, 5)
axis_label_size = 18
tick_label_size = 14
spine_width = 1
tick_width = 1
tick_length = 5

# Requested colors: 0deg blue, 45deg red
geo_styles = {
    "0deg": {"color": "#0b7bcb", "label": "0°"},   # blue
    "45deg": {"color": "#ca1a1a", "label": "45°"}, # red
}

# -----------------------------
# Load graph dictionary
# -----------------------------
if "GL_final" in globals():
    GL_use = GL_final
else:
    p = GRAPH_LABELED_PATH
    with open(p, "rb") as f:
        GL_use = pickle.load(f)

def get_graphs(GL_dict, geom, graph_view="core", use_all=True, sim_index=0):
    graphs = GL_dict[geom][graph_view]
    if use_all:
        return graphs
    if len(graphs) == 0:
        return []
    idx = int(np.clip(sim_index, 0, len(graphs) - 1))
    return [graphs[idx]]

# -----------------------------
# Collect degree data + stats
# -----------------------------
pooled_degrees = {}
stats = {}

for geom in GEOMS:
    if geom not in GL_use:
        print(f"Warning: {geom} not found.")
        continue

    pooled = []
    per_graph_var = []
    per_graph_mean = []

    for G in get_graphs(GL_use, geom, GRAPH_VIEW, USE_ALL_SIMS, SIM_INDEX):
        deg = np.asarray([d for _, d in G.degree()], dtype=float)
        if deg.size == 0:
            continue

        pooled.extend(deg.tolist())
        per_graph_mean.append(float(np.mean(deg)))
        per_graph_var.append(float(np.var(deg, ddof=0)))

    pooled = np.asarray(pooled, dtype=float)
    pg_var = np.asarray(per_graph_var, dtype=float)
    pg_mean = np.asarray(per_graph_mean, dtype=float)

    pooled_degrees[geom] = pooled
    stats[geom] = {
        "n_graphs": int(pg_var.size),
        "n_nodes_total": int(pooled.size),
        "pooled_mean_degree": float(np.mean(pooled)) if pooled.size else np.nan,
        "pooled_variance_degree": float(np.var(pooled, ddof=0)) if pooled.size else np.nan,
        "mean_per_graph_degree": float(np.mean(pg_mean)) if pg_mean.size else np.nan,
        "mean_per_graph_variance": float(np.mean(pg_var)) if pg_var.size else np.nan,
        "sd_per_graph_variance": float(np.std(pg_var, ddof=1)) if pg_var.size > 1 else 0.0,
    }

# -----------------------------
# Plot histogram (0° vs 45°)
# -----------------------------
fig, ax = plt.subplots(figsize=figsize)

for geom in GEOMS:
    if geom not in pooled_degrees:
        continue

    deg = pooled_degrees[geom]
    if deg.size == 0:
        continue

    c = geo_styles[geom]["color"]
    lbl = geo_styles[geom]["label"]

    if HIST_MODE == "percentage":
        weights = np.ones_like(deg) * (100.0 / deg.size)
        ax.hist(
            deg,
            bins=BIN_EDGES,
            weights=weights,
            histtype="stepfilled",
            color=c,
            alpha=0.28,
            linewidth=0,
        )
        ax.hist(
            deg,
            bins=BIN_EDGES,
            weights=weights,
            histtype="step",
            color=c,
            linewidth=2.2,
            label=lbl,
        )
        ylabel = "Percentage (%)"

    elif HIST_MODE == "pdf":
        ax.hist(
            deg,
            bins=BIN_EDGES,
            density=True,
            histtype="stepfilled",
            color=c,
            alpha=0.28,
            linewidth=0,
        )
        ax.hist(
            deg,
            bins=BIN_EDGES,
            density=True,
            histtype="step",
            color=c,
            linewidth=2.2,
            label=lbl,
        )
        ylabel = "PDF"
    else:
        raise ValueError("HIST_MODE must be 'percentage' or 'pdf'")

ax.set_xticks(XTICKS)
ax.set_xlabel("Degree", fontsize=axis_label_size)
ax.set_ylabel(ylabel, fontsize=axis_label_size)
ax.set_title(f"Degree distribution: 0° vs 45° ({GRAPH_VIEW})", fontsize=axis_label_size)
ax.tick_params(axis="both", which="major", labelsize=tick_label_size, width=tick_width, length=tick_length)
for sp in ax.spines.values():
    sp.set_linewidth(spine_width)
ax.legend(frameon=False, fontsize=tick_label_size)
fig.tight_layout()
plt.show()

# -----------------------------
# Print requested statistics
# -----------------------------
print("\nDegree statistics (0° and 45°)")
for geom in GEOMS:
    if geom not in stats:
        continue
    s = stats[geom]
    print(
        f"{geom:>5s} | "
        f"pooled mean={s['pooled_mean_degree']:.6f}, "
        f"pooled variance={s['pooled_variance_degree']:.6f}, "
        f"mean per-graph variance={s['mean_per_graph_variance']:.6f}, "
        f"SD per-graph variance={s['sd_per_graph_variance']:.6f}, "
        f"n_graphs={s['n_graphs']}"
    )
 

from scipy.stats import ttest_ind, mannwhitneyu

# Extract per-graph variance arrays
var_0 = np.array([
    np.var([d for _, d in G.degree()], ddof=0)
    for G in get_graphs(GL_use, "0deg", GRAPH_VIEW, True, SIM_INDEX)
])

var_45 = np.array([
    np.var([d for _, d in G.degree()], ddof=0)
    for G in get_graphs(GL_use, "45deg", GRAPH_VIEW, True, SIM_INDEX)
])

# --- t-test ---
t_stat, p_t = ttest_ind(var_0, var_45, equal_var=False)

# --- Mann-Whitney (safer) ---
u_stat, p_u = mannwhitneyu(var_0, var_45, alternative="two-sided")

# --- effect size (Cohen's d) ---
def cohen_d(a, b):
    return (np.mean(a) - np.mean(b)) / np.sqrt((np.var(a) + np.var(b)) / 2)

d = cohen_d(var_0, var_45)

print("\nStatistical comparison (variance):")
print(f"t-test p-value       = {p_t:.4e}")
print(f"Mann-Whitney p-value = {p_u:.4e}")
print(f"Cohen's d            = {d:.4f}")   
plt.figure(figsize=(6,4))
plt.boxplot([var_0, var_45], labels=["0°", "45°"])
plt.ylabel("Degree variance")
plt.title("Per-graph degree variance")
plt.show()
```

```python
# Variance box plot for all four geometries (degree variance per graph)

import numpy as np
import matplotlib.pyplot as plt
import pickle
from pathlib import Path

# -----------------------------
# Config
# -----------------------------
GRAPH_VIEW = "core"  # "core" or "full"
GEOMS = ["0deg", "15deg", "30deg", "45deg"]

# Colors (0 blue, 45 red as requested)
geom_colors = {
    "0deg": "#1f77b4",   # blue
    "15deg": "#ff7f0e",  # orange
    "30deg": "#2ca02c",  # green
    "45deg": "#d62728",  # red
}

plt.rcParams["font.family"] = "Times New Roman"

# -----------------------------
# Load graph dictionary
# -----------------------------
if "GL_prop" in globals():
    GL_use = GL_prop
elif "GL_final" in globals():
    GL_use = GL_final
else:
    p = GRAPH_LABELED_PATH
    with open(p, "rb") as f:
        GL_use = pickle.load(f)

# -----------------------------
# Collect per-graph degree variance
# -----------------------------
var_by_geom = {}

for geom in GEOMS:
    if geom not in GL_use:
        print(f"Warning: {geom} not found. Skipping.")
        continue

    vars_this_geom = []
    graphs = GL_use[geom][GRAPH_VIEW]

    for G in graphs:
        deg = np.asarray([d for _, d in G.degree()], dtype=float)
        if deg.size == 0:
            continue
        vars_this_geom.append(float(np.var(deg, ddof=0)))  # population variance per graph

    var_by_geom[geom] = np.asarray(vars_this_geom, dtype=float)

# Keep only geometries with data
plot_geoms = [g for g in GEOMS if g in var_by_geom and var_by_geom[g].size > 0]
if not plot_geoms:
    raise ValueError("No valid variance data found for selected geometries.")

data = [var_by_geom[g] for g in plot_geoms]
labels = [g.replace("deg", "°") for g in plot_geoms]
colors = [geom_colors[g] for g in plot_geoms]

# -----------------------------
# Box plot
# -----------------------------
fig, ax = plt.subplots(figsize=(8, 5.5))

bp = ax.boxplot(
    data,
    labels=labels,
    patch_artist=True,
    showmeans=True,
    meanline=False,
    whis=1.5,
)

for patch, c in zip(bp["boxes"], colors):
    patch.set_facecolor(c)
    patch.set_alpha(0.28)
    patch.set_edgecolor(c)
    patch.set_linewidth(2.0)

for whisker in bp["whiskers"]:
    whisker.set_linewidth(1.5)
for cap in bp["caps"]:
    cap.set_linewidth(1.5)
for med in bp["medians"]:
    med.set_color("black")
    med.set_linewidth(2.0)
for mean in bp["means"]:
    mean.set_marker("o")
    mean.set_markerfacecolor("white")
    mean.set_markeredgecolor("black")
    mean.set_markersize(6)

ax.set_xlabel("Geometry", fontsize=18)
ax.set_ylabel("Per-Graph Degree Variance", fontsize=18)
ax.set_title(f"Degree Variance Box Plot ({GRAPH_VIEW})", fontsize=18)
ax.tick_params(axis="both", which="major", labelsize=14, width=1, length=5)
for sp in ax.spines.values():
    sp.set_linewidth(1)

fig.tight_layout()
plt.show()

# -----------------------------
# Numeric summary
# -----------------------------
print("\nPer-graph degree variance summary:")
for g in plot_geoms:
    v = var_by_geom[g]
    print(
        f"{g:>5s} | n_graphs={len(v):2d} | "
        f"mean={np.mean(v):.6f} | std={np.std(v, ddof=1) if len(v) > 1 else 0.0:.6f} | "
        f"median={np.median(v):.6f}"
    )
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd

# --------------------------------------------------
# Fixed 4x4 layout with labels A-P
# --------------------------------------------------
labels = {
    0: "A", 1: "B", 2: "C", 3: "D",
    4: "E", 5: "F", 6: "G", 7: "H",
    8: "I", 9: "J", 10: "K", 11: "L",
    12: "M", 13: "N", 14: "O", 15: "P"
}

# positions chosen so all three graphs are directly comparable
pos = {
    0: (0, 3), 1: (1, 3), 2: (2, 3), 3: (3, 3),
    4: (0, 2), 5: (1, 2), 6: (2, 2), 7: (3, 2),
    8: (0, 1), 9: (1, 1), 10: (2, 1), 11: (3, 1),
    12: (0, 0), 13: (1, 0), 14: (2, 0), 15: (3, 0)
}

# --------------------------------------------------
# Stage 1: baseline 4x4 grid
# --------------------------------------------------
G1 = nx.grid_2d_graph(4, 4)
G1 = nx.convert_node_labels_to_integers(G1, ordering="sorted")

# --------------------------------------------------
# Stage 2: from Stage 1
# remove BC, FG, GK
# add EJ, GL, KN, LO
# --------------------------------------------------
G2 = G1.copy()
G2.remove_edges_from([
    (1, 2),   # B-C
    (5, 6),   # F-G
    # (6, 10),  # G-K
    (9, 13), # J-N
    
])
G2.add_edges_from([
    (4, 9),   # E-J
    (6, 11),  # G-L
    (10, 13), # K-N
    (11, 14), # L-O
])

# --------------------------------------------------
# Stage 3: from Stage 2
# additionally remove AB, CD
# --------------------------------------------------
G3 = G2.copy()
G3.remove_edges_from([
    (0, 1),   # A-B
    (2, 3),   # C-D
    # (13, 14), # N-O
    (13, 10), # N-K
])

# --------------------------------------------------
# Metric calculation
# --------------------------------------------------
def compute_metrics(G):
    degree_dict = dict(G.degree())
    closeness_dict = nx.closeness_centrality(G)
    betweenness_dict = nx.betweenness_centrality(G, normalized=True)
    cluster_dict = nx.cluster.clustering(G)

    mean_degree = sum(degree_dict.values()) / G.number_of_nodes()
    mean_closeness = sum(closeness_dict.values()) / G.number_of_nodes()
    mean_betweenness = sum(betweenness_dict.values()) / G.number_of_nodes()
    mean_cluster = sum(cluster_dict.values()) / G.number_of_nodes()

    return {
        "num_nodes": G.number_of_nodes(),
        "num_edges": G.number_of_edges(),
        "mean_degree": mean_degree,
        "mean_closeness": mean_closeness,
        "mean_betweenness": mean_betweenness,
        "mean_cluster": mean_cluster
    }

results = pd.DataFrame([
    {"stage": "Stage 1", **compute_metrics(G1)},
    {"stage": "Stage 2", **compute_metrics(G2)},
    {"stage": "Stage 3", **compute_metrics(G3)},
])

# nicer display
pd.set_option("display.precision", 4)
print(results)

# --------------------------------------------------
# Plotting
# --------------------------------------------------
def draw_graph(ax, G, title):
    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=700)
    nx.draw_networkx_edges(G, pos, ax=ax, width=1.8)
    nx.draw_networkx_labels(G, pos, labels=labels, ax=ax, font_size=11)
    ax.set_title(title, fontsize=12)
    ax.set_axis_off()

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
draw_graph(axes[0], G1, "Stage 1: Baseline")
draw_graph(axes[1], G2, "Stage 2: Degree increases")
draw_graph(axes[2], G3, "Stage 3: Degree decreases")
plt.tight_layout()
plt.show()
```
